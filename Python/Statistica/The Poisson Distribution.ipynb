<h1>Derivation of the Poisson Distribution</h1>

<p>The binomial distribution has another useful approximation in the case in which $p << 1$ (when the probability of success is small) and $N >> 1$ (the number of tries is really large). In this case, the number of positive outcomes is much smaller than the number of tries, $n << N$, and the factorial function can be approximated as</p>

\begin{equation}
N! = N (N - 1)... (N - n + 1) (N - n)! \approx N^n (N - n)!
\end{equation}

<p>We are also interested in finding an approximation for the $q^n$ term that appears in the binomial. For this we set</p>

\begin{equation}
\ln q^{N - n} = \ln (1 - p)^{N - n} = (N - n) \ln (1 - p) \approx -p (N - n)
\end{equation}

<p>You can in fact see for yourself that the function $y = -x$ accurately approximates the function $ln (1 - x)$ for $x << 1$:</p>

```python
import numpy as np
import plotly.graph_objects as go

x = np.arange(-3, 1, 0.01)
y1 = -x
y2 = np.log(1 - x)

fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=-x, name='y = -x'))
fig.add_trace(go.Scatter(x=x, y=np.log(1 - x), name='y = log(1 - x)'))
fig.update_xaxes(range=[-3, 3], zeroline=True, zerolinewidth=1, zerolinecolor='black')
fig.update_yaxes(range=[-1, 1], zeroline=True, zerolinewidth=1, zerolinecolor='black')
fig.update_yaxes(scaleanchor = "x", scaleratio = 1)
fig.update_layout(template='plotly_white')
fig.show()
```

<p>As $N >> n$, $N - n \approx N$ and thus $-p (N - n) \approx -p N$ and therefore we obtain the approximation</p>

\begin{equation}
q^{N - n} \approx e^{-pN}
\end{equation}

<p>These approximations can be used into the equation for the Binomial distribution</p>

\begin{equation}
  P(n) = \frac{N!}{(N - n)! n!} p^n \cdot q^{N - n}
\end{equation}

<p>to give</p>

\begin{equation}
  P(n) = \frac{N^n (N - n)!}{(N - n)! n!} p^n \cdot e^{-pN} = \frac{N^n}{n!} p^n \cdot e^{-pN} = \frac{(pN)^n}{n!} e^{-pN}
\end{equation}
