<h1>Discrete Colors in Python</h1>
How to use and configure discrete color sequences, also known as categorical or qualitative color scales.

## <h3 id="discrete-vs-continuous-color">Discrete vs Continuous Color</h3>

<p>In the same way as the X or Y position of a mark in cartesian coordinates can be used to represent continuous values (i.e. amounts or moments in time) or categories (i.e. labels), color can be used to represent continuous or discrete data. This page is about using color to represent <strong>categorical</strong> data using discrete colors, but Plotly can also <a href="https://plotly.com/python/colorscales/">represent continuous values with color</a>.</p>

## <h3 id="discrete-color-concepts">Discrete Color Concepts</h3>

<p>This document explains the following discrete-color-related concepts:</p>

<ul>

<li><strong>color sequences</strong> are lists of colors to be mapped onto discrete data values. No interpolation occurs when using color sequences, unlike with <a href="https://plotly.com/python/colorscales/">continuous color scales</a>, and each color is used as-is. Color sequence defaults depend on the <code>layout.colorway</code> attribute of the active <a href="https://plotly.com/python/templates/">template</a>, and can be explicitly specified using the <code>color_discrete_sequence</code> argument for many <a href="https://plotly.com/python/plotly-express/">Plotly Express</a> functions.</li>

<li><strong>legends</strong> are visible representations of the mapping between colors and data values. Legend markers also change shape when used with various kinds of traces, such as symbols or lines for scatter-like traces. <a href="https://plotly.com/python/legend/">Legends are configurable</a> under the <code>layout.legend</code> attribute. Legends are the discrete equivalent of <a href="https://plotly.com/python/colorscales/">continuous color bars</a></li>

</ul>

## <h3 id="discrete-color-with-plotly-express">Discrete Color with Plotly Express</h3>

<p>Most Plotly Express functions accept a <code>color</code> argument which automatically assigns data values to discrete colors <strong>if the data is non-numeric</strong>. If the data is numeric, the color will automatically be considered <a href="https://plotly.com/python/colorscales/">continuous</a>. This means that numeric strings must be parsed to be used for continuous color, and conversely, numbers used as category codes must be converted to strings.</p>

## <h3 id="color-sequences-in-plotly-express">Color Sequences in Plotly Express</h3>

<p>By default, Plotly Express will use the color sequence from the active <a href="https://plotly.com/python/templates/">template</a>'s <code>layout.colorway</code> attribute, and the default active template is <code>plotly</code> which uses the <code>plotly</code> color sequence. You can choose any of the following built-in qualitative color sequences from the <code>px.colors.qualitative</code> module, however, or define your own.</p>

```python
import plotly.express as px

fig = px.colors.qualitative.swatches()
fig.show()
```

<p>Color sequences in the <code>px.colors.qualitative</code> module are stored as lists of CSS colors:</p>

```python
import plotly.express as px

print(px.colors.qualitative.Plotly)
```

<p>Here is an example that creates a scatter plot using Plotly Express, with points colored using the built-in qualitative <code>G10</code> color sequence.</p>

```python
import plotly.express as px
df = px.data.gapminder()
fig = px.line(df, y="lifeExp", x="year", color="continent", line_group="country",
              line_shape="spline", render_mode="svg",
             color_discrete_sequence=px.colors.qualitative.G10,
             title="Built-in G10 color sequence")

fig.show()
```
