<h1>Plotly Express in Python</h1>
Plotly Express is a terse, consistent, high-level API for creating figures.


## <h3 id="overview">Overview</h3>

<p>The <code>plotly.express</code> module (usually imported as <code>px</code>) contains functions that can create entire figures at once, and is referred to as Plotly Express or PX. Plotly Express is a built-in part of the <code>plotly</code> library, and is the recommended starting point for creating most common figures. Every Plotly Express function uses <a href="/python/graph-objects/">graph objects</a> internally and returns a <code>plotly.graph_objects.Figure</code> instance. Throughout the <code>plotly</code> documentation, you will find the Plotly Express way of building figures at the top of any applicable page, followed by a section on how to use graph objects to build similar figures. Any figure created in a single function call with Plotly Express could be created using graph objects alone, but with between 5 and 100 times more code.</p>

<p>Plotly Express provides <a href="https://plotly.com/python-api-reference/plotly.express.html">more than 30 functions for creating different types of figures</a>. The API for these functions was carefully designed to be as consistent and easy to learn as possible, making it easy to switch from a scatter plot to a bar chart to a histogram to a sunburst chart throughout a data exploration session. <em>Scroll down for a gallery of Plotly Express plots, each made in a single function call.</em></p>

## <h3 id="highlevel-features">High-Level Features</h3>

<p>The Plotly Express API in general offers the following features:</p>

<ul>

<li><strong>A single entry point into <code>plotly</code></strong>: just <code>import plotly.express as px</code> and get access to <a href="https://plotly.com/python-api-reference/plotly.express.html">all the plotting functions</a>, plus <a href="https://plotly.com/python-api-reference/generated/plotly.data.html#module-plotly.data">built-in demo datasets under <code>px.data</code></a> and <a href="https://plotly.com/python-api-reference/generated/plotly.colors.html#module-plotly.colors">built-in color scales and sequences under <code>px.color</code></a>. Every PX function returns a <code>plotly.graph_objects.Figure</code> object, so you can edit it using all the same methods like <a href="https://plotly.com/python/creating-and-updating-figures/#updating-figures"><code>update_layout</code> and <code>add_trace</code></a>.</li>

<li><strong>Sensible, Overridable Defaults</strong>: PX functions will infer sensible defaults wherever possible, and will always let you override them.</li>

<li><strong>Flexible Input Formats</strong>: PX functions <a href="https://plotly.com/python/px-arguments/">accept input in a variety of formats</a>, from <code>list</code>s and <code>dict</code>s to <a href="https://plotly.com/python/wide-form/">long-form or wide-form <code>DataFrame</code>s</a> to <a href="https://plotly.com/python/imshow/"><code>numpy</code> arrays and <code>xarrays</code></a> to <a href="https://plotly.com/python/maps/">GeoPandas <code>GeoDataFrames</code></a>.</li>

<li><strong>Automatic Trace and Layout configuration</strong>: PX functions will create one <a href="https://plotly.com/python/figure-structure">trace</a> per animation frame for each unique combination of data values mapped to discrete color, symbol, line-dash, facet-row and/or facet-column. Traces' <a href="https://plotly.com/python/legend/"><code>legendgroup</code> and <code>showlegend</code> attributes</a> are set such that only one legend item appears per unique combination of discrete color, symbol and/or line-dash. Traces are automatically linked to a correctly-configured <a href="https://plotly.com/python/figure-structure">subplot of the appropriate type</a>.</li>

<li><strong>Automatic Figure Labelling</strong>: PX functions <a href="https://plotly.com/python/figure-labels/">label axes, legends and colorbars</a> based in the input <code>DataFrame</code> or <code>xarray</code>, and provide <a href="https://plotly.com/python/styling-plotly-express/">extra control with the <code>labels</code> argument</a>.</li>

<li><strong>Automatic Hover Labels</strong>: PX functions populate the hover-label using the labels mentioned above, and provide <a href="https://plotly.com/python/hover-text-and-formatting/">extra control with the <code>hover_name</code> and <code>hover_data</code> arguments</a>.</li>

<li><strong>Styling Control</strong>: PX functions <a href="https://plotly.com/python/styling-plotly-express/">read styling information from the default figure template</a>, and support commonly-needed <a href="https://plotly.com/python/styling-plotly-express/">cosmetic controls like <code>category_orders</code> and <code>color_discrete_map</code></a> to precisely control categorical variables.</li>

<li><strong>Uniform Color Handling</strong>: PX functions automatically switch between <a href="https://plotly.com/python/colorscales/">continuous</a> and <a href="https://plotly.com/python/discrete-color/">categorical color</a> based on the input type.</li>

<li><strong>Faceting</strong>: the 2D-cartesian plotting functions support <a href="https://plotly.com/python/facet-plots/">row, column and wrapped facetting with <code>facet_row</code>, <code>facet_col</code> and <code>facet_col_wrap</code> arguments</a>.</li>

<li><strong>Marginal Plots</strong>: the 2D-cartesian plotting functions support <a href="https://plotly.com/python/marginal-plots/">marginal distribution plots</a> with the <code>marginal</code>, <code>marginal_x</code> and <code>marginal_y</code> arguments.</li>

<li><strong>A Pandas backend</strong>: the 2D-cartesian plotting functions are available as <a href="https://plotly.com/python/pandas-backend/">a Pandas plotting backend</a> so you can call them via <code>df.plot()</code>.</li>

<li><strong>Trendlines</strong>: <code>px.scatter</code> supports <a href="https://plotly.com/python/linear-fits/">built-in trendlines with accessible model output</a>.</li>

<li><strong>Animations</strong>: many PX functions support <a href="https://plotly.com/python/animations/">simple animation support via the <code>animation_frame</code> and <code>animation_group</code> arguments</a>.</li>

<li><strong>Automatic WebGL switching</strong>: for sufficiently large scatter plots, PX will automatically <a href="https://plotly.com/python/webgl-vs-svg/">use WebGL for hardware-accelerated rendering</a>.</li>

</ul>

## <h3 id="gallery">Gallery</h3>

### <h4 id="scatter-line-area-and-bar-charts">Scatter, Line, Area and Bar Charts</h4>

<p><strong>Read more about <a href="https://plotly.com/python/line-and-scatter/">scatter plots</a> and <a href="https://plotly.com/python/discrete-color/">discrete color</a>.</strong></p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species")
fig.show()
```

<p><strong>Read more about <a href="https://plotly.com/python/linear-fits/">trendlines</a> and <a href="https://plotly.com/python/templates/">templates</a> and <a href="https://plotly.com/python/marginal-plots/">marginal distribution plots</a>.</strong></p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", marginal_y="violin",
           marginal_x="box", trendline="ols", template="simple_white")
fig.show()
```

<p><strong>Read more about <a href="https://plotly.com/python/error-bars/">error bars</a>.</strong></p>

```python
import plotly.express as px
df = px.data.iris()
df["e"] = df["sepal_width"]/100
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", error_x="e", error_y="e")
fig.show()
```

```python
<p><strong>Read more about <a href="https://plotly.com/python/bar-charts/">bar charts</a>.</strong></p>
```

```python
import plotly.express as px
df = px.data.tips()
fig = px.bar(df, x="sex", y="total_bill", color="smoker", barmode="group")
fig.show()
```

```python
import plotly.express as px
df = px.data.medals_long()

fig = px.bar(df, x="medal", y="count", color="nation",
             pattern_shape="nation", pattern_shape_sequence=[".", "x", "+"])
fig.show()
```

<p><strong>Read more about <a href="https://plotly.com/python/facet-plots/">facet plots</a>.</strong></p>

```python
import plotly.express as px
df = px.data.tips()
fig = px.bar(df, x="sex", y="total_bill", color="smoker", barmode="group", facet_row="time", facet_col="day",
       category_orders={"day": ["Thur", "Fri", "Sat", "Sun"], "time": ["Lunch", "Dinner"]})
fig.show()
```

<p><strong>Read more about <a href="https://plotly.com/python/splom/">scatterplot matrices (SPLOMs)</a>.</strong></p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter_matrix(df, dimensions=["sepal_width", "sepal_length", "petal_width", "petal_length"], color="species")
fig.show()
```
