<h1>Line Charts in Python</h1>
How to make line charts in Python with Plotly. Examples on creating and styling line charts in Python with Plotly.

## <h3 id="line-plots-with-plotlyexpress">Line Plots with plotly.express</h3>

<p><a href="https://plotly.com/python/plotly-express/">Plotly Express</a> is the easy-to-use, high-level interface to Plotly, which <a href="https://plotly.com/python/px-arguments/">operates on a variety of types of data</a> and produces <a href="https://plotly.com/python/styling-plotly-express/">easy-to-style figures</a>.
With <code>px.line</code>, each data point is represented as a vertex (which location is given by the <code>x</code> and <code>y</code> columns) of a <strong>polyline mark</strong> in 2D space.</p>

<p>For more examples of line plots, see the <a href="https://plotly.com/python/line-and-scatter/">line and scatter notebook</a>.</p>

```python
import plotly.express as px

df = px.data.gapminder().query("country=='Canada'")
fig = px.line(df, x="year", y="lifeExp", title='Life expectancy in Canada')
fig.show()
```

## <h3 id="line-plots-with-column-encoding-color">Line Plots with column encoding color</h3>

```python
import plotly.express as px

df = px.data.gapminder().query("continent=='Oceania'")
fig = px.line(df, x="year", y="lifeExp", color='country')
fig.show()
```

## <h3 id="line-charts-with-markers">Line charts with markers</h3>

<p>The <code>markers</code> argument can be set to <code>True</code> to show markers on lines.</p>

```python
import plotly.express as px
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country', markers=True)
fig.show()
```

<p>The <code>symbol</code> argument can be used to map a data field to the marker symbol. A <a href="https://plotly.com/python/marker-style/">wide variety of symbols</a> are available.</p>

```python
import plotly.express as px
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country', symbol="country")
fig.show()
```

## <h3 id="line-plots-on-date-axes">Line plots on Date axes</h3>

<p>Line plots can be made on using any type of cartesian axis, including <a href="https://plotly.com/python/axes/">linear</a>, <a href="https://plotly.com/python/log-plot/">logarithmic</a>, <a href="https://plotly.com/python/categorical-axes/">categorical</a> or date axes. Line plots on date axes are often called <a href="https://plotly.com/python/time-series/">time-series charts</a>.</p>

<p>Plotly auto-sets the axis type to a date format when the corresponding data are either ISO-formatted date strings or if they're a <a href="https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html">date pandas column</a> or <a href="https://docs.scipy.org/doc/numpy/reference/arrays.datetime.html">datetime NumPy array</a>.</p>

```python
import plotly.express as px

df = px.data.stocks()
fig = px.line(df, x='date', y="GOOG")
fig.show()
```

## <h3 id="line-plot-with-goscatter">Line Plot with go.Scatter</h3>

<p>If Plotly Express does not provide a good starting point, it is possible to use <a href="https://plotly.com/python/graph-objects/">the more generic <code>go.Scatter</code> class from <code>plotly.graph_objects</code></a>. Whereas <code>plotly.express</code> has two functions <code>scatter</code> and <code>line</code>, <code>go.Scatter</code> can be used both for plotting points (makers) or lines, depending on the value of <code>mode</code>. The different options of <code>go.Scatter</code> are documented in its <a href="https://plotly.com/python/reference/scatter/">reference page</a>.</p>

### <h4 id="simple-line-plot">Simple Line Plot</h4>

```python
import plotly.graph_objects as go
import numpy as np

x = np.arange(10)

fig = go.Figure(data=go.Scatter(x=x, y=x**2))
fig.show()
```

### <h4 id="line-plot-modes">Line Plot Modes</h4>

```python
import plotly.graph_objects as go

# Create random data with numpy
import numpy as np
np.random.seed(1)

N = 100
random_x = np.linspace(0, 1, N)
random_y0 = np.random.randn(N) + 5
random_y1 = np.random.randn(N)
random_y2 = np.random.randn(N) - 5

# Create traces
fig = go.Figure()
fig.add_trace(go.Scatter(x=random_x, y=random_y0,
                    mode='lines',
                    name='lines'))
fig.add_trace(go.Scatter(x=random_x, y=random_y1,
                    mode='lines+markers',
                    name='lines+markers'))
fig.add_trace(go.Scatter(x=random_x, y=random_y2,
                    mode='markers', name='markers'))

fig.show()
```
