<h1>Scatter Plots in Python</h1>
How to make scatter plots in Python with Plotly.

# <h2 id="scatter-plots-with-plotly-express">Scatter plots with Plotly Express</h2>

<p><a href="https://plotly.com/python/plotly-express/">Plotly Express</a> is the easy-to-use, high-level interface to Plotly, which <a href="https://plotly.com/python/px-arguments/">operates on a variety of types of data</a> and produces <a href="https://plotly.com/python/styling-plotly-express/">easy-to-style figures</a>.</p>

<p>With <code>px.scatter</code>, each data point is represented as a marker point, whose location is given by the <code>x</code> and <code>y</code> columns.</p>

```python
# x and y given as array_like objects
import plotly.express as px
fig = px.scatter(x=[0, 1, 2, 3, 4], y=[0, 1, 4, 9, 16])
fig.show()
```

```python
# x and y given as DataFrame columns
import plotly.express as px
df = px.data.iris() # iris is a pandas DataFrame
fig = px.scatter(df, x="sepal_width", y="sepal_length")
fig.show()
```

### <h4 id="setting-size-and-color-with-column-names">Setting size and color with column names</h4>

<p>Scatter plots with variable-sized circular markers are often known as <a href="https://plotly.com/python/bubble-charts/">bubble charts</a>. Note that <code>color</code> and <code>size</code> data are added to hover information. You can add other columns to hover data with the <code>hover_data</code> argument of <code>px.scatter</code>.</p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", size='petal_length', hover_data=['petal_width'])
fig.show()
```

<p>Color can be <a href="https://plotly.com/python/colorscales/">continuous</a> as follows, or <a href="https://plotly.com/python/discrete-color/">discrete/categorical</a> as above.</p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color='petal_length')
fig.show()
```

<p>The <code>symbol</code> argument can be mapped to a column as well. A <a href="https://plotly.com/python/marker-style/">wide variety of symbols</a> are available.</p>

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", symbol="species")
fig.show()
```

## <h3 id="facetting">Facetting</h3>

<p>Scatter plots support <a href="https://plotly.com/python/facet-plots/">faceting</a>.</p>

```python
import plotly.express as px
df = px.data.tips()
fig = px.scatter(df, x="total_bill", y="tip", color="smoker", facet_col="sex", facet_row="time")
fig.show()
```

## <h3 id="linear-regression-and-other-trendlines">Linear Regression and Other Trendlines</h3>

<p>Scatter plots support <a href="https://plotly.com/python/linear-fits/">linear and non-linear trendlines</a>.</p>

```python
import plotly.express as px

df = px.data.tips()
fig = px.scatter(df, x="total_bill", y="tip", trendline="ols")
fig.show()
```

# <h2 id="line-plots-with-plotly-express">Line plots with Plotly Express</h2>

```python
import plotly.express as px
import numpy as np

t = np.linspace(0, 2*np.pi, 100)

fig = px.line(x=t, y=np.cos(t), labels={'x':'t', 'y':'cos(t)'})
fig.show()
```

```python
import plotly.express as px
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country')
fig.show()
```

<p>The <code>markers</code> argument can be set to <code>True</code> to show markers on lines.</p>

```python
import plotly.express as px
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country', markers=True)
fig.show()
```

<p>The <code>symbol</code> argument can be used to map a data field to the marker symbol. A <a href="https://plotly.com/python/marker-style/">wide variety of symbols</a> are available.</p>

```python
import plotly.express as px
df = px.data.gapminder().query("continent == 'Oceania'")
fig = px.line(df, x='year', y='lifeExp', color='country', symbol="country")
fig.show()
```

## <h3 id="line-plots-on-date-axes">Line plots on Date axes</h3>

<p>Line plots can be made on using any type of cartesian axis, including <a href="https://plotly.com/python/axes/">linear</a>, <a href="https://plotly.com/python/log-plot/">logarithmic</a>, <a href="https://plotly.com/python/categorical-axes/">categorical</a> or date axes. Line plots on date axes are often called <a href="https://plotly.com/python/time-series/">time-series charts</a>.</p>

<p>Plotly auto-sets the axis type to a date format when the corresponding data are either ISO-formatted date strings or if they're a <a href="https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html">date pandas column</a> or <a href="https://docs.scipy.org/doc/numpy/reference/arrays.datetime.html">datetime NumPy array</a>.</p>

```python
import plotly.express as px

df = px.data.stocks()
fig = px.line(df, x='date', y="GOOG")
fig.show()
```
