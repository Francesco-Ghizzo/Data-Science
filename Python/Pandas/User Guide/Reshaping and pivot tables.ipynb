## <h3><a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot.html" title="pandas.pivot"><code>pivot()</code></a></h3>

<p>Data is often stored in so-called “stacked” or “record” format. In a “record” or “wide” format, typically there is one row for each subject. In the “stacked” or “long” format there are multiple rows for each subject where applicable.</p>

```python
data = {
   "value": range(12),
   "variable": ["A"] * 3 + ["B"] * 3 + ["C"] * 3 + ["D"] * 3,
   "date": pd.to_datetime(["2020-01-03", "2020-01-04", "2020-01-05"] * 4)
}

df = pd.DataFrame(data)
```

<p>To perform time series operations with each unique variable, a better representation would be where the <code>columns</code> are the unique variables and an <code>index</code> of dates identifies individual observations. To reshape the data into this form, we use the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot.html" title="pandas.DataFrame.pivot"><code>DataFrame.pivot()</code></a> method (also implemented as a top level function <a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot.html" title="pandas.pivot"><code>pivot()</code></a>):</p>

```python
pivoted = df.pivot(index="date", columns="variable", values="value")
```

```python
pivoted
```

<p>If the <code>values</code> argument is omitted, and the input <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> has more than one column of values which are not used as column or index inputs to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot.html" title="pandas.DataFrame.pivot"><code>pivot()</code></a>, then the resulting “pivoted” <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> will have <a href="https://pandas.pydata.org/docs/user_guide/advanced.html#advanced-hierarchical">hierarchical columns</a> whose topmost level indicates the respective value column:</p>

```python
df["value2"] = df["value"] * 2
```

```python
pivoted = df.pivot(index="date", columns="variable")
```

```python
pivoted
```

<p>You can then select subsets from the pivoted <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a>:</p>

```python
pivoted["value2"]
```

<p>Note that this returns a view on the underlying data in the case where the data are homogeneously-typed.</p>

<p class="admonition-title">Note</p>
<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot.html" title="pandas.pivot"><code>pivot()</code></a> can only handle unique rows specified by <code>index</code> and <code>columns</code>.
If you data contains duplicates, use <a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html" title="pandas.pivot_table"><code>pivot_table()</code></a>.</p>

## <h3><a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html" title="pandas.pivot_table"><code>pivot_table()</code></a></h3>

<p>While <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot.html" title="pandas.DataFrame.pivot"><code>pivot()</code></a> provides general purpose pivoting with various data types, pandas also provides <a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html" title="pandas.pivot_table"><code>pivot_table()</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html" title="pandas.DataFrame.pivot_table"><code>pivot_table()</code></a> for pivoting with aggregation of numeric data.</p>

<p>The function <a href="https://pandas.pydata.org/docs/reference/api/pandas.pivot_table.html" title="pandas.pivot_table"><code>pivot_table()</code></a> can be used to create spreadsheet-style pivot tables. See the <a href="cookbook.html#cookbook-pivot">cookbook</a> for some advanced strategies.</p>

```python
import datetime
```

```python
df = pd.DataFrame(
    {
        "A": ["one", "one", "two", "three"] * 6,
        "B": ["A", "B", "C"] * 8,
        "C": ["foo", "foo", "foo", "bar", "bar", "bar"] * 4,
        "D": np.random.randn(24),
        "E": np.random.randn(24),
        "F": [datetime.datetime(2013, i, 1) for i in range(1, 13)]
        + [datetime.datetime(2013, i, 15) for i in range(1, 13)],
    }
)
```

```python
df
```

```python
pd.pivot_table(df, values="D", index=["A", "B"], columns="C")
```

```python
pd.pivot_table(
    df, values=["D", "E"],
    index=["B"],
    columns=["A", "C"],
    aggfunc="sum",
)
```

<p>Also, you can use <a href="https://pandas.pydata.org/docs/reference/api/pandas.Grouper.html" title="pandas.Grouper"><code>Grouper</code></a> for <code>index</code> and <code>columns</code> keywords. For detail of <a href="https://pandas.pydata.org/docs/reference/api/pandas.Grouper.html" title="pandas.Grouper"><code>Grouper</code></a>, see <a href="https://pandas.pydata.org/docs/user_guide/groupby.html#groupby-specify">Grouping with a Grouper specification</a>.</p>

```python
pd.pivot_table(df, values="D", index=pd.Grouper(freq="M", key="F"), columns="C")
```

<p>The result is a <a href="../reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> potentially having a <a href="../reference/api/pandas.MultiIndex.html#pandas.MultiIndex" title="pandas.MultiIndex"><code>MultiIndex</code></a> on the index or column. If the <code>values</code> column name is not given, the pivot table will include all of the data in an additional level of hierarchy in the columns:</p>

```python
pd.pivot_table(df[["A", "B", "C", "D", "E"]], index=["A", "B"], columns=["C"])
```

### <h4>Adding margins</h4>

<p>Passing <code>margins=True</code> to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot_table.html" title="pandas.DataFrame.pivot_table"><code>pivot_table()</code></a> will add a row and column with an <code>All</code> label with partial group aggregates across the categories on the rows and columns:</p>

```python
table = df.pivot_table(
    index=["A", "B"],
    columns="C",
    values=["D", "E"],
    margins=True,
    aggfunc="std"
)
```

```python
table
```

<p>Additionally, you can call <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>DataFrame.stack()</code></a> to display a pivoted DataFrame as having a multi-level index:</p>

```python
table.stack(future_stack=True)
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html" title="pandas.DataFrame.unstack"><code>unstack()</code></h2>

<img src="https://pandas.pydata.org/docs/_images/reshaping_stack.png">

<p>Closely related to the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot.html" title="pandas.DataFrame.pivot"><code>pivot()</code></a> method are the related <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html" title="pandas.DataFrame.unstack"><code>unstack()</code></a> methods available on <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a>. These methods are designed to work together with <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html" title="pandas.MultiIndex"><code>MultiIndex</code></a> objects (see the section on <a href="advanced.html#advanced-hierarchical">hierarchical indexing</a>).</p>

<ul>
<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a>: “pivot” a level of the (possibly hierarchical) column labels, returning a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> with an index with a new inner-most level of row labels.</p></li>
<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html" title="pandas.DataFrame.unstack"><code>unstack()</code></a>: (inverse operation of <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a>) “pivot” a level of the (possibly hierarchical) row index to the column axis, producing a reshaped <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> with a new inner-most level of column labels.</p></li>
</ul>

<img src="https://pandas.pydata.org/docs/_images/reshaping_unstack.png">

```python
tuples = [
   ["bar", "bar", "baz", "baz", "foo", "foo", "qux", "qux"],
   ["one", "two", "one", "two", "one", "two", "one", "two"],
]
```

```python
index = pd.MultiIndex.from_arrays(tuples, names=["first", "second"])
```

```python
df = pd.DataFrame(np.random.randn(8, 2), index=index, columns=["A", "B"])
```

```python
df2 = df[:4]
```

```python
df2
```

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> function “compresses” a level in the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> columns to produce either:</p>

<ul>
<li><p>A <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a>, in the case of a <a href="https://pandas.pydata.org/docs/reference/api/pandas.Index.html" title="pandas.Index"><code>Index</code></a> in the columns.</p></li>
<li><p>A <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a>, in the case of a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html" title="pandas.MultiIndex"><code>MultiIndex</code></a> in the columns.</p></li>
</ul>

<p>If the columns have a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html" title="pandas.MultiIndex"><code>MultiIndex</code></a>, you can choose which level to stack. The stacked level becomes the new lowest level in a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html" title="pandas.MultiIndex"><code>MultiIndex</code></a> on the columns:</p>

```python
stacked = df2.stack(future_stack=True)
```

```python
stacked
```

<p>With a “stacked” <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> (having a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html" title="pandas.MultiIndex"><code>MultiIndex</code></a> as the <code>index</code>), the inverse operation of <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> is <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html" title="pandas.DataFrame.unstack"><code>unstack()</code></a>, which by default unstacks the <strong>last level</strong>:</p>

```python
stacked.unstack()
```

```python
stacked.unstack(1)
```

```python
stacked.unstack(0)
```

<img id="reshaping-unstack-by-name" src="https://pandas.pydata.org/docs/_images/reshaping_unstack_1.png">

<p>If the indexes have names, you can use the level names instead of specifying the level numbers:</p>

```python
stacked.unstack("second")
```

<img src="https://pandas.pydata.org/docs/_images/reshaping_unstack_0.png">

<p>Notice that the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html#pandas.DataFrame.unstack" title="pandas.DataFrame.unstack"><code>unstack()</code></a> methods implicitly sort the index levels involved. Hence a call to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.stack.html" title="pandas.DataFrame.stack"><code>stack()</code></a> and then <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.unstack.html" title="pandas.DataFrame.unstack"><code>unstack()</code></a>, or vice versa, will result in a <strong>sorted</strong> copy of the original <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html"><code>Series</code></a>:</p>

```python
index = pd.MultiIndex.from_product([[2, 1], ["a", "b"]])
```

```python
df = pd.DataFrame(np.random.randn(4), index=index, columns=["A"])
```

```python
df
```

```python
all(df.unstack().stack(future_stack=True) == df.sort_index())
```

## <h3>Multiple levels</h3>

<p>You may also stack or unstack more than one level at a time by passing a list of levels, in which case the end result is as if each level in the list were processed individually.</p>

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("A", "cat", "long"),
        ("B", "cat", "long"),
        ("A", "dog", "short"),
        ("B", "dog", "short"),
    ],
    names=["exp", "animal", "hair_length"],
)
```

```python
df = pd.DataFrame(np.random.randn(4, 4), columns=columns)
```

```python
df
```

```python
df.stack(level=["animal", "hair_length"], future_stack=True)
```

<p>The list of levels can contain either level names or level numbers but not a mixture of the two.</p>

```python
# df.stack(level=['animal', 'hair_length'], future_stack=True)
# from above is equivalent to:
df.stack(level=[1, 2], future_stack=True)
```

## <h3>Missing data</h3>

<p>Unstacking can result in missing values if subgroups do not have the same set of labels. By default, missing values will be replaced with the default fill value for that data type.</p>

```python
columns = pd.MultiIndex.from_tuples(
    [
        ("A", "cat"),
        ("B", "dog"),
        ("B", "cat"),
        ("A", "dog"),
    ],
    names=["exp", "animal"],
)
```

```python
index = pd.MultiIndex.from_product(
    [("bar", "baz", "foo", "qux"), ("one", "two")], names=["first", "second"]
)
```

```python
df = pd.DataFrame(np.random.randn(8, 4), index=index, columns=columns)
```

```python
df3 = df.iloc[[0, 1, 4, 7], [1, 2]]
```

```python
df3
```

```python
df3.unstack()
```

<p>The missing value can be filled with a specific value with the <code>fill_value</code> argument.</p>

```python
df3.unstack(fill_value=-1e9)
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.melt.html" title="pandas.melt"><code>melt()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.wide_to_long.html" title="pandas.wide_to_long"><code>wide_to_long()</code></a></h2>

<img src="https://pandas.pydata.org/docs/_images/reshaping_melt.png">

<p>The top-level <a href="https://pandas.pydata.org/docs/reference/api/pandas.melt.html" title="pandas.melt"><code>melt()</code></a> function and the corresponding <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.melt.html" title="pandas.DataFrame.melt"><code>DataFrame.melt()</code></a> are useful to massage a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> into a format where one or more columns are <em>identifier variables</em>, while all other columns, considered <em>measured variables</em>, are “unpivoted” to the row axis, leaving just two non-identifier columns, “variable” and “value”. The names of those columns can be customized by supplying the <code>var_name</code> and <code>value_name</code> parameters.</p>

```python
cheese = pd.DataFrame(
    {
        "first": ["John", "Mary"],
        "last": ["Doe", "Bo"],
        "height": [5.5, 6.0],
        "weight": [130, 150],
    }
)
```

```python
cheese
```

```python
cheese.melt(id_vars=["first", "last"])
```

```python
cheese.melt(id_vars=["first", "last"], var_name="quantity")
```

<p>When transforming a DataFrame using <a href="https://pandas.pydata.org/docs/reference/api/pandas.melt.html" title="pandas.melt"><code>melt()</code></a>, the index will be ignored.
The original index values can be kept by setting the <code>ignore_index=False</code> parameter to <code>False</code> (default is <code>True</code>).
<code>ignore_index=False</code> will however duplicate index values.</p>

```python
index = pd.MultiIndex.from_tuples([("person", "A"), ("person", "B")])
```

```python
cheese = pd.DataFrame(
    {
        "first": ["John", "Mary"],
        "last": ["Doe", "Bo"],
        "height": [5.5, 6.0],
        "weight": [130, 150],
    },
    index=index,
)
```

```python
cheese
```

```python
cheese.melt(id_vars=["first", "last"])
```

```python
cheese.melt(id_vars=["first", "last"], ignore_index=False)
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.wide_to_long.html" title="pandas.wide_to_long"><code>wide_to_long()</code></a> is similar to <a href="https://pandas.pydata.org/docs/reference/api/pandas.melt.html" title="pandas.melt"><code>melt()</code></a> with more customization for column matching.</p>

```python
dft = pd.DataFrame(
    {
        "A1970": {0: "a", 1: "b", 2: "c"},
        "A1980": {0: "d", 1: "e", 2: "f"},
        "B1970": {0: 2.5, 1: 1.2, 2: 0.7},
        "B1980": {0: 3.2, 1: 1.3, 2: 0.1},
        "X": dict(zip(range(3), np.random.randn(3))),
    }
)
```

```python
dft["id"] = dft.index
```

```python
dft
```

```python
pd.wide_to_long(dft, ["A", "B"], i="id", j="year")
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.get_dummies.html" title="pandas.get_dummies"><code>get_dummies()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.from_dummies.html" title="pandas.from_dummies"><code>from_dummies()</code></a></h2>

<p>To convert categorical variables of a <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> into a “dummy” or “indicator”, <a href="https://pandas.pydata.org/docs/reference/api/pandas.get_dummies.html" title="pandas.get_dummies"><code>get_dummies()</code></a> creates a new <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> with columns of the unique variables and the values representing the presence of those variables per row.</p>

```python
df = pd.DataFrame({"key": list("bbacab"), "data1": range(6)})
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a></h2>

<p>Use <a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a> to compute a cross-tabulation of two (or more) factors. By default <a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a> computes a frequency table of the factors unless an array of values and an aggregation function are passed.</p>

<p>Any <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> passed will have their name attributes used unless row or column names for the cross-tabulation are specified</p>

```python
a = np.array(["foo", "foo", "bar", "bar", "foo", "foo"], dtype=object)
```

```python
b = np.array(["one", "one", "two", "one", "two", "one"], dtype=object)
```

```python
c = np.array(["dull", "dull", "shiny", "dull", "dull", "shiny"], dtype=object)
```

```python
pd.crosstab(a, [b, c], rownames=["a"], colnames=["b", "c"])
```

<p>If <a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a> receives only two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a>, it will provide a frequency table.</p>

```python
df = pd.DataFrame(
    {"A": [1, 2, 2, 2, 2], "B": [3, 3, 4, 4, 4], "C": [1, 1, np.nan, 1, 1]}
)
```

```python
df
```

```python
pd.crosstab(df["A"], df["B"])
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a> can also summarize to <a href="https://pandas.pydata.org/docs/reference/api/pandas.Categorical.html" title="pandas.Categorical"><code>Categorical</code></a> data.</p>

```python
foo = pd.Categorical(["a", "b"], categories=["a", "b", "c"])
```

```python
bar = pd.Categorical(["d", "e"], categories=["d", "e", "f"])
```

```python
pd.crosstab(foo, bar)
```

<p>For <a href="https://pandas.pydata.org/docs/reference/api/pandas.Categorical.html" title="pandas.Categorical"><code>Categorical</code></a> data, to include <strong>all</strong> of data categories even if the actual data does not contain any instances of a particular category, use <code>dropna=False</code>.</p>

```python
pd.crosstab(foo, bar, dropna=False)
```

## <h3>Normalization</h3>

<p>Frequency tables can also be normalized to show percentages rather than counts using the <code>normalize</code> argument:</p>

```python
pd.crosstab(df["A"], df["B"], normalize=True)
```

<p><code>normalize</code> can also normalize values within each row or within each column:</p>

```python
pd.crosstab(df["A"], df["B"], normalize="columns")
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html" title="pandas.crosstab"><code>crosstab()</code></a> can also accept a third <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> and an aggregation function (<code>aggfunc</code>) that will be applied to the values of the third <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> within each group defined by the first two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a>:</p>

```python
pd.crosstab(df["A"], df["B"], values=df["C"], aggfunc="sum")
```

## <h3>Adding margins</h3>

<p><code>margins=True</code> will add a row and column with an <code>All</code> label with partial group aggregates across the categories on the rows and columns:</p>

```python
pd.crosstab(
    df["A"], df["B"], values=df["C"], aggfunc="sum", normalize=True, margins=True
)
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.cut.html" title="pandas.cut"><code>cut()</code></a></h2>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.cut.html" title="pandas.cut"><code>cut()</code></a> function computes groupings for the values of the input array and is often used to transform continuous variables to discrete or categorical variables:</p>

<p>An integer <code>bins</code> will form equal-width bins.</p>

```python
ages = np.array([10, 15, 13, 12, 23, 25, 28, 59, 60])
```

```python
pd.cut(ages, bins=3)
```

<p>A list of ordered bin edges will assign an interval for each variable.</p>

```python
pd.cut(ages, bins=[0, 18, 35, 70])
```

<p>If the <code>bins</code> keyword is an <a href="https://pandas.pydata.org/docs/reference/api/pandas.IntervalIndex.html" title="pandas.IntervalIndex"><code>IntervalIndex</code></a>, then these will be used to bin the passed data.</p>

```python
pd.cut(ages, bins=pd.IntervalIndex.from_breaks([0, 40, 70]))
```
