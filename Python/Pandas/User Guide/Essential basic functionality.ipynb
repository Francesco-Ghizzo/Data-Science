## <h3>Comparing if objects are equivalent</h3>

<p>Often you may find that there is more than one way to compute the same result.  As a simple example, consider <code>df + df</code> and <code>df * 2</code>. To test
that these two computations produce the same result, given the tools
shown above, you might imagine using <code>(df + df == df * 2).all()</code>. But in
fact, this expression is False:</p>

```python
df + df == df * 2
```

```python
(df + df == df * 2).all()
```

<p>Notice that the boolean DataFrame <code>df + df == df * 2</code> contains some False values!
This is because NaNs do not compare as equals:</p>

```python
np.nan == np.nan
```

<p>So, NDFrames (such as Series and DataFrames)
have an <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.equals.html#pandas.DataFrame.equals" title="pandas.DataFrame.equals"><code>equals()</code></a> method for testing equality, with NaNs in corresponding locations treated as equal.</p>

```python
(df + df).equals(df * 2)
```

<p>Note that the Series or DataFrame index needs to be in the same order for equality to be True:</p>

```python
df1 = pd.DataFrame({"col": ["foo", 0, np.nan]})

df2 = pd.DataFrame({"col": [np.nan, 0, "foo"]}, index=[2, 1, 0])
```

```python
df1.equals(df2)
```

```python
df1.equals(df2.sort_index())
```

## <h3>Combining overlapping data sets</h3>

<p>A problem occasionally arising is the combination of two similar data sets where values in one are preferred over the other. An example would be two data series representing a particular economic indicator where one is considered to be of “higher quality”.
However, the lower quality series might extend further back in history or have more complete data coverage. As such, we would like to combine two DataFrame objects where missing values in one DataFrame are conditionally filled with like-labeled values from the other DataFrame.
The function implementing this operation is <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.combine_first.html#pandas.DataFrame.combine_first" title="pandas.DataFrame.combine_first"><code>combine_first()</code></a>, which we illustrate:</p>

```python
df1 = pd.DataFrame(
    {"A": [1.0, np.nan, 3.0, 5.0, np.nan], "B": [np.nan, 2.0, 3.0, np.nan, 6.0]}
)
```

```python
df2 = pd.DataFrame(
    {
        "A": [5.0, 2.0, 4.0, np.nan, 3.0, 7.0],
        "B": [np.nan, np.nan, 3.0, 4.0, 6.0, 8.0],
    }
)
```

```python
df1
```

```python
df2
```

```python
df1.combine_first(df2)
```

## <h3>General DataFrame combine</h3>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.combine_first.html#pandas.DataFrame.combine_first" title="pandas.DataFrame.combine_first"><code>combine_first()</code></a> method above calls the more general <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.combine.html#pandas.DataFrame.combine" title="pandas.DataFrame.combine"><code>DataFrame.combine()</code></a>.
This method takes another DataFrame and a combiner function, aligns the input DataFrame and then passes the combiner function pairs of Series (i.e., columns whose names are the same).</p>

<p>So, for instance, to reproduce <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.combine_first.html#pandas.DataFrame.combine_first" title="pandas.DataFrame.combine_first"><code>combine_first()</code></a> as above:</p>

```python
def combiner(x, y):
  return np.where(pd.isna(x), y, x)

df1.combine(df2, combiner)
```

# <h2>Descriptive statistics</h2>

<p>There exists a large number of methods for computing descriptive statistics and other related operations on <a href="https://pandas.pydata.org/docs/reference/series.html#api-series-stats">Series</a>, <a href="https://pandas.pydata.org/docs/reference/frame.html#api-dataframe-stats">DataFrame</a>.
Most of these are aggregations (hence producing a lower-dimensional result) like <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sum.html#pandas.DataFrame.sum" title="pandas.DataFrame.sum"><code>sum()</code></a>, <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.mean.html#pandas.DataFrame.mean" title="pandas.DataFrame.mean"><code>mean()</code></a>, and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.quantile.html#pandas.DataFrame.quantile" title="pandas.DataFrame.quantile"><code>quantile()</code></a>, but some of them, like <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cumsum.html#pandas.DataFrame.cumsum" title="pandas.DataFrame.cumsum"><code>cumsum()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cumprod.html#pandas.DataFrame.cumprod" title="pandas.DataFrame.cumprod"><code>cumprod()</code></a>, produce an object of the same size.
Generally speaking, these methods take an <strong>axis</strong> argument, just like <em>ndarray.{sum, std, …}</em>, but the axis can be specified by name or integer:</p>

<ul>
<li><p><strong>Series</strong>: no axis argument needed</p></li>
<li><p><strong>DataFrame</strong>: “index” (axis=0, default), “columns” (axis=1)</p></li>
</ul>

<p>For example:</p>

```python
df
```

```python
df.mean(0)
```

```python
df.mean(1)
```

<p>All such methods have a <code>skipna</code> option signaling whether to exclude missing data (<code>True</code> by default):</p>

```python
df.sum(0, skipna=False)
```

```python
df.sum(axis=1, skipna=True)
```

<p>Combined with the broadcasting / arithmetic behavior, one can describe various statistical procedures, like standardization (rendering data zero mean and standard deviation of 1), very concisely:</p>

```python
ts_stand = (df - df.mean()) / df.std()

ts_stand.std()
```

```python
xs_stand = df.sub(df.mean(1), axis=0).div(df.std(1), axis=0)

xs_stand.std(1)
```

<p>Note that methods like <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cumsum.html#pandas.DataFrame.cumsum" title="pandas.DataFrame.cumsum"><code>cumsum()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cumprod.html#pandas.DataFrame.cumprod" title="pandas.DataFrame.cumprod"><code>cumprod()</code></a> preserve the location of <code>NaN</code> values. This is somewhat different from <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.expanding.html#pandas.DataFrame.expanding" title="pandas.DataFrame.expanding"><code>expanding()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.rolling.html#pandas.DataFrame.rolling" title="pandas.DataFrame.rolling"><code>rolling()</code></a> since <code>NaN</code> behavior is furthermore dictated by a <code>min_periods</code> parameter.</p>

```python
df.cumsum()
```

## <h3>Summarizing data: describe</h3>

<p>There is a convenient <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html#pandas.DataFrame.describe" title="pandas.DataFrame.describe"><code>describe()</code></a> function which computes a variety of summary statistics about a Series or the columns of a DataFrame (excluding NAs of course):</p>

```python
series = pd.Series(np.random.randn(1000))

series[::2] = np.nan

series.describe()
```

```python
frame = pd.DataFrame(np.random.randn(1000, 5), columns=["a", "b", "c", "d", "e"])

frame.iloc[::2] = np.nan

frame.describe()
```

<p>You can select specific percentiles to include in the output:</p>

```python
series.describe(percentiles=[0.05, 0.25, 0.75, 0.95])
```

<p>By default, the median is always included.</p>
<p>For a non-numerical Series object, <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html#pandas.DataFrame.describe" title="pandas.Series.describe"><code>describe()</code></a> will give a simple summary of the number of unique values and most frequently occurring values:</p>

```python
s = pd.Series(["a", "a", "b", "b", "a", "a", np.nan, "c", "d", "a"])

s.describe()
```

<p>Note that on a mixed-type DataFrame object, <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html#pandas.DataFrame.describe" title="pandas.DataFrame.describe"><code>describe()</code></a> will restrict the summary to include only numerical columns or, if none are, only categorical columns:</p>

```python
frame = pd.DataFrame({"a": ["Yes", "Yes", "No", "No"], "b": range(4)})

frame.describe()
```

<p>This behavior can be controlled by providing a list of types as <code>include</code>/<code>exclude</code>
arguments. The special value <code>all</code> can also be used:</p>

```python
frame.describe(include=["object"])
```

```python
frame.describe(include=["number"])
```

```python
frame.describe(include="all")
```

<p>That feature relies on <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.select_dtypes.html#pandas.DataFrame.select_dtypes">select_dtypes</a>. Refer to there for details about accepted inputs.</p>

## <h3>Discretization and quantiling</h3>

<p>Continuous values can be discretized using the <a href="https://pandas.pydata.org/docs/reference/api/pandas.cut.html#pandas.cut" title="pandas.cut"><code>cut()</code></a> (bins based on values) and <a href="https://pandas.pydata.org/docs/reference/api/pandas.qcut.html#pandas.qcut" title="pandas.qcut"><code>qcut()</code></a> (bins based on sample quantiles) functions:</p>

```python
arr = np.random.randn(20)

factor = pd.cut(arr, 4)
```

```python
factor
```

```python
factor = pd.cut(arr, [-5, -1, 0, 1, 5])
```

```python
factor
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.qcut.html#pandas.qcut" title="pandas.qcut"><code>qcut()</code></a> computes sample quantiles. For example, we could slice up some normally distributed data into equal-size quartiles like so:</p>

```python
arr = np.random.randn(30)

factor = pd.qcut(arr, [0, 0.25, 0.5, 0.75, 1])
```

```python
factor
```

```python
factor
```

## <h3>Tablewise function application</h3>

<p><code>DataFrames</code> and <code>Series</code> can be passed into functions.
However, if the function needs to be called in a chain, consider using the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pipe.html#pandas.DataFrame.pipe" title="pandas.DataFrame.pipe"><code>pipe()</code></a> method.</p>

<p>First some setup:</p>

```python
def extract_city_name(df):
       """
       Chicago, IL -> Chicago for city_name column
       """
       df["city_name"] = df["city_and_code"].str.split(",").str.get(0)
       return df
```

```python
def add_country_name(df, country_name=None):
       """
       Chicago -> Chicago-US for city_name column
       """
       col = "city_name"
       df["city_and_country"] = df[col] + country_name
       return df
```

```python
df_p = pd.DataFrame({"city_and_code": ["Chicago, IL"]})
```

<p><code>extract_city_name</code> and <code>add_country_name</code> are functions taking and returning <code>DataFrames</code>.</p>

<p>Now compare the following:</p>

```python
add_country_name(extract_city_name(df_p), country_name="US")
```

<p>Is equivalent to:</p>

```python
df_p.pipe(extract_city_name).pipe(add_country_name, country_name="US")
```

<p>pandas encourages the second style, which is known as method chaining.
<code>pipe</code> makes it easy to use your own or another library’s functions
in method chains, alongside pandas’ methods.</p>
<p>In the example above, the functions <code>extract_city_name</code> and <code>add_country_name</code> each expected a <code>DataFrame</code> as the first positional argument.
What if the function you wish to apply takes its data as, say, the second argument?
In this case, provide <code>pipe</code> with a tuple of <code>(callable, data_keyword)</code>.
<code>.pipe</code> will route the <code>DataFrame</code> to the argument specified in the tuple.</p>
<p>For example, we can fit a regression using statsmodels. Their API expects a formula first and a <code>DataFrame</code> as the second argument, <code>data</code>. We pass in the function, keyword pair <code>(sm.ols, 'data')</code> to <code>pipe</code>:</p>

```python
import statsmodels.formula.api as sm

bb = pd.read_csv("https://raw.githubusercontent.com/pandas-dev/pandas/refs/heads/main/doc/data/baseball.csv", index_col="id")

(
       bb.query("h > 0")
       .assign(ln_h=lambda df: np.log(df.h))
       .pipe((sm.ols, "data"), "hr ~ ln_h + year + g + C(lg)")
       .fit()
       .summary()
   )
```

<p>The pipe method is inspired by unix pipes and more recently <a href="https://github.com/tidyverse/dplyr">dplyr</a> and <a href="https://github.com/tidyverse/magrittr">magrittr</a>, which
have introduced the popular <code>(%&gt;%)</code> (read pipe) operator for <a href="https://www.r-project.org">R</a>.
The implementation of <code>pipe</code> here is quite clean and feels right at home in Python.
We encourage you to view the source code of <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pipe.html#pandas.DataFrame.pipe" title="pandas.DataFrame.pipe"><code>pipe()</code></a>.</p>

## <h3>Row or column-wise function application</h3>

<p>Arbitrary functions can be applied along the axes of a DataFrame
using the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> method, which, like the descriptive
statistics methods, takes an optional <code>axis</code> argument:</p>

```python
df.apply(lambda x: np.mean(x))
```

```python
df.apply(lambda x: np.mean(x), axis=1)
```

```python
df.apply(lambda x: x.max() - x.min())
```

```python
df.apply(np.cumsum)
```

```python
df.apply(np.exp)
```

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> method will also dispatch on a string method name.</p>

```python
df.apply("mean")
```

```python
df.apply("mean", axis=1)
```

<p>The return type of the function passed to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> affects the
type of the final output from <code>DataFrame.apply</code> for the default behaviour:</p>
<ul>
<li><p>If the applied function returns a <code>Series</code>, the final output is a <code>DataFrame</code>.
The columns match the index of the <code>Series</code> returned by the applied function.</p></li>
<li><p>If the applied function returns any other type, the final output is a <code>Series</code>.</p></li>
</ul>
<p>This default behaviour can be overridden using the <code>result_type</code>, which
accepts three options: <code>reduce</code>, <code>broadcast</code>, and <code>expand</code>.
These will determine how list-likes return values expand (or not) to a <code>DataFrame</code>.</p>
<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> combined with some cleverness can be used to answer many questions
about a data set. For example, suppose we wanted to extract the date where the
maximum value for each column occurred:</p>

```python
tsdf = pd.DataFrame(
       np.random.randn(1000, 3),
       columns=["A", "B", "C"],
       index=pd.date_range("1/1/2000", periods=1000),
   )

tsdf.apply(lambda x: x.idxmax())
```

<p>You may also pass additional arguments and keyword arguments to the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> method.</p>

```python
def subtract_and_divide(x, sub, divide=1):
       return (x - sub) / divide

df_udf = pd.DataFrame(np.ones((2, 2)))

df_udf.apply(subtract_and_divide, sub=5, divide=3)
```

When set to True, the passed function will instead receive an ndarray object, which has positive performance implications if you do not need the indexing functionality.</p>

## <h3>Aggregation API</h3>

<p>The aggregation API allows one to express possibly multiple aggregation operations in a single concise way.
This API is similar across pandas objects, see <a href="https://pandas.pydata.org/docs/user_guide/groupby.html#groupby-aggregate">groupby API</a>, the <a href="https://pandas.pydata.org/docs/user_guide/window.html#window-overview">window API</a>, and the <a href="https://pandas.pydata.org/docs/user_guide/timeseries.html#timeseries-aggregate">resample API</a>.
The entry point for aggregation is <a href=".https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.aggregate.html#pandas.DataFrame.aggregate" title="pandas.DataFrame.aggregate"><code>DataFrame.aggregate()</code></a>, or the alias <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.agg.html#pandas.DataFrame.agg" title="pandas.DataFrame.agg"><code>DataFrame.agg()</code></a>.</p>

<p>We will use a similar starting frame from above:</p>

```python
tsdf = pd.DataFrame(
       np.random.randn(10, 3),
       columns=["A", "B", "C"],
       index=pd.date_range("1/1/2000", periods=10),
   )

tsdf.iloc[3:7] = np.nan
```

```python
tsdf
```

<p>Using a single function is equivalent to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a>. You can also pass named methods as strings. These will return a <code>Series</code> of the aggregated output:</p>

```python
tsdf.agg(lambda x: np.sum(x))
```

```python
tsdf.agg("sum")
```

```python
# these are equivalent to a .sum() because we are aggregating
# on a single function
tsdf.sum()
```

# <h2>Reindexing and altering labels</h2>

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.reindex.html#pandas.Series.reindex" title="pandas.Series.reindex"><code>reindex()</code></a> is the fundamental data alignment method in pandas.
It is used to implement nearly all other features relying on label-alignment functionality. To <em>reindex</em> means to conform the data to match a given set of labels along a particular axis. This accomplishes several things:</p>

<ul>
<li><p>Reorders the existing data to match a new set of labels</p></li>
<li><p>Inserts missing value (NA) markers in label locations where no data for
that label existed</p></li>
<li><p>If specified, <strong>fill</strong> data for missing labels using logic (highly relevant
to working with time series data)</p></li>
</ul>

<p>Here is a simple example:</p>

```python
s = pd.Series(np.random.randn(5), index=["a", "b", "c", "d", "e"])
```

```python
s
```

```python
s.reindex(["e", "b", "f", "d"])
```

<p>Here, the <code>f</code> label was not contained in the Series and hence appears as <code>NaN</code> in the result.</p>

<p>With a DataFrame, you can simultaneously reindex the index and columns:</p>

```python
df
```

```python
df.reindex(index=["c", "f", "b"], columns=["three", "two", "one"])
```

<p>Note that the <code>Index</code> objects containing the actual axis labels can be <strong>shared</strong> between objects. So if we have a Series and a DataFrame, the following can be done:</p>

```python
rs = s.reindex(df.index)
```

```python
rs
```

```python
rs.index is df.index
```

<p>This means that the reindexed Series’s index is the same Python object as the DataFrame’s index.</p>

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reindex.html#pandas.DataFrame.reindex" title="pandas.DataFrame.reindex"><code>DataFrame.reindex()</code></a> also supports an “axis-style” calling convention, where you specify a single <code>labels</code> argument and the <code>axis</code> it applies to.</p>

```python
df.reindex(["c", "f", "b"], axis="index")
```

```python
df.reindex(["three", "two", "one"], axis="columns")
```

<p>See also</p>
<p><a href="https://pandas.pydata.org/docs/user_guide/advanced.html#advanced">MultiIndex / Advanced Indexing</a> is an even more concise way of doing reindexing.</p>
</div>

<div class="alert alert-block alert-info">
<p>Note</p>
<p>When writing performance-sensitive code, there is a good reason to spend some time becoming a reindexing ninja: <strong>many operations are faster on pre-aligned data</strong>. # <h2>Iteration</h2>

<p>The behavior of basic iteration over pandas objects depends on the type. When iterating over a Series, it is regarded as array-like, and basic iteration produces the values. DataFrames follow the dict-like convention of iterating over the “keys” of the objects.</p>

<p>In short, basic iteration (<code>for i in object</code>) produces:</p>

<ul>
<li><p><strong>Series</strong>: values</p></li>
<li><p><strong>DataFrame</strong>: column labels</p></li>
</ul>

<p>Thus, for example, iterating over a DataFrame gives you the column names:</p>

```python
df = pd.DataFrame(
       {"col1": np.random.randn(3), "col2": np.random.randn(3)}, index=["a", "b", "c"]
   )
```

```python
for col in df:
    print(col)
```

<p>pandas objects also have the dict-like <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.items.html" title="pandas.DataFrame.items"><code>items()</code></a> method to iterate over the (key, value) pairs.</p>

<p>To iterate over the rows of a DataFrame, you can use the following methods:</p>

<ul>
<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iterrows.html" title="pandas.DataFrame.iterrows"><code>iterrows()</code></a>: Iterate over the rows of a DataFrame as (index, Series) pairs.
This converts the rows to Series objects, which can change the dtypes and has some performance implications.</p></li>
<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.itertuples.html" title="pandas.DataFrame.itertuples"><code>itertuples()</code></a>: Iterate over the rows of a DataFrame as namedtuples of the values.  This is a lot faster than <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iterrows.html" title="pandas.DataFrame.iterrows"><code>iterrows()</code></a>, and is in most cases preferable to use to iterate over the values of a DataFrame.</p></li>
</ul>

<div class="alert alert-block alert-warning">
<p>Warning</p>

<p>Iterating through pandas objects is generally <strong>slow</strong>. In many cases, iterating manually over the rows is not needed and can be avoided with one of the following approaches:</p>

<ul>
<li><p>Look for a <em>vectorized</em> solution: many operations can be performed using built-in methods or NumPy functions, (boolean) indexing, …</p></li>
<li><p>When you have a function that cannot work on the full DataFrame/Series at once, it is better to use <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.apply.html#pandas.DataFrame.apply" title="pandas.DataFrame.apply"><code>apply()</code></a> instead of iterating over the values. See the docs on <a href="https://pandas.pydata.org/docs/user_guide/basics.html#basics-apply">function application</a>.</p></li>
<li><p>If you need to do iterative manipulations on the values but performance is important, consider writing the inner loop with cython or numba. See the <a href="https://pandas.pydata.org/docs/user_guide/enhancingperf.html#enhancingperf">enhancing performance</a> section for some examples of this approach.</p></li>
</ul>
</div>

<div class="alert alert-block alert-warning">
<p>Warning</p>
<p>You should <strong>never modify</strong> something you are iterating over. This is not guaranteed to work in all cases. Depending on the
data types, the iterator returns a copy and not a view, and writing to it will have no effect!</p>

<p>For example, in the following case setting the value has no effect:</p>

```python
df = pd.DataFrame({"a": [1, 2, 3], "b": ["a", "b", "c"]})
```

```python
for index, row in df.iterrows():
    row["a"] = 10
```

```python
df
```

## <h3>iterrows</h3>

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iterrows.html#pandas.DataFrame.iterrows" title="pandas.DataFrame.iterrows"><code>iterrows()</code></a> allows you to iterate through the rows of a DataFrame as Series objects. It returns an iterator yielding each index value along with a Series containing the data in each row:</p>

```python
for row_index, row in df.iterrows():
    print(row_index, row, sep="\n")
```

<div class="alert alert-block alert-info">
<p>Note</p>
<p>Because <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iterrows.html#pandas.DataFrame.iterrows" title="pandas.DataFrame.iterrows"><code>iterrows()</code></a> returns a Series for each row, it does <strong>not</strong> preserve dtypes across the rows (dtypes are preserved across columns for DataFrames). For example,</p>

```python
df_orig = pd.DataFrame([[1, 1.5]], columns=["int", "float"])
```

```python
df_orig.dtypes
```

```python
row = next(df_orig.iterrows())[1]
```

```python
row
```

<p>All values in <code>row</code>, returned as a Series, are now upcasted to floats, also the original integer value in column <code>x</code>:</p>

```python
row["int"].dtype
```

```python
df_orig["int"].dtype
```

<p>To preserve dtypes while iterating over the rows, it is better to use <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.itertuples.html" title="pandas.DataFrame.itertuples"><code>itertuples()</code></a> which returns namedtuples of the values and which is generally much faster than <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iterrows.html" title="pandas.DataFrame.iterrows"><code>iterrows()</code></a>.</p>

<p>For instance, a contrived way to transpose the DataFrame would be:</p>

```python
df2 = pd.DataFrame({"x": [1, 2, 3], "y": [4, 5, 6]})
```

```python
print(df2)
```

## <h3>itertuples</h3>

The first element of the tuple will be the row’s corresponding index value, while the remaining values are the row values.</p>

<p>For instance:</p>

```python
for row in df.itertuples():
    print(row)
```

<p>This method does not convert the row to a Series object; it merely returns the values inside a namedtuple. Therefore,

<div class="alert alert-block alert-info">
<p>Note</p>
<p>The column names will be renamed to positional names if they are invalid Python identifiers, repeated, or start with an underscore.
With a large number of columns (&gt;255), regular tuples are returned.</p>

# <h2>Vectorized string methods</h2>

<p>Series is equipped with a set of string processing methods that make it easy to operate on each element of the array. Perhaps most importantly, these methods exclude missing/NA values automatically. These are accessed via the Series’s <code>str</code> attribute and generally have names matching the equivalent (scalar) built-in string methods. For example:</p>
<blockquote>

# <h2>Sorting</h2>

<p>pandas supports three kinds of sorting: sorting by index labels,
sorting by column values, and sorting by a combination of both.</p>

## <h3>By index</h3>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.sort_index.html" title="pandas.Series.sort_index"><code>Series.sort_index()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_index.html" title="pandas.DataFrame.sort_index"><code>DataFrame.sort_index()</code></a> methods are used to sort a pandas object by its index levels.</p>

```python
df = pd.DataFrame(
       {
           "one": pd.Series(np.random.randn(3), index=["a", "b", "c"]),
           "two": pd.Series(np.random.randn(4), index=["a", "b", "c", "d"]),
           "three": pd.Series(np.random.randn(3), index=["b", "c", "d"]),
       }
   )
```

```python
unsorted_df = df.reindex(
       index=["a", "d", "c", "b"], columns=["three", "two", "one"]
   )

unsorted_df
```

```python
# DataFrame
unsorted_df.sort_index()
```

```python
unsorted_df.sort_index(ascending=False)
```

```python
unsorted_df.sort_index(axis=1)
```

```python
# Series
unsorted_df["three"].sort_index()
```

<p id="basics-sort-index-key">Sorting by index also supports a <code>key</code> parameter that takes a callable
function to apply to the index being sorted. For <code>MultiIndex</code> objects, the key is applied per-level to the levels specified by <code>level</code>.</p>

```python
s1 = pd.DataFrame({"a": ["B", "a", "C"], "b": [1, 2, 3], "c": [2, 3, 4]}).set_index(
       list("ab")
   )

s1
```

```python
s1.sort_index(level="a")
```

```python
s1.sort_index(level="a", key=lambda idx: idx.str.lower())
```

<p>For information on key sorting by value, see <a href="https://pandas.pydata.org/docs/user_guide/basics.html#basics-sort-value-key">value sorting</a>.</p>

## <h3>By values</h3>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.sort_values.html#pandas.Series.sort_values" title="pandas.Series.sort_values"><code>Series.sort_values()</code></a> method is used to sort a <code>Series</code> by its values. The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html#pandas.DataFrame.sort_values" title="pandas.DataFrame.sort_values"><code>DataFrame.sort_values()</code></a> method is used to sort a <code>DataFrame</code> by its column or row values.
The optional <code>by</code> parameter to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html#pandas.DataFrame.sort_values" title="pandas.DataFrame.sort_values"><code>DataFrame.sort_values()</code></a> may used to specify one or more columns to use to determine the sorted order.</p>

```python
df1 = pd.DataFrame(
       {"one": [2, 1, 1, 1], "two": [1, 3, 2, 4], "three": [5, 4, 3, 2]}
   )

df1.sort_values(by="two")
```

<p>The <code>by</code> parameter can take a list of column names, e.g.:</p>

```python
df1[["one", "two", "three"]].sort_values(by=["one", "two"])
```

<p>These methods have special treatment of NA values via the <code>na_position</code> argument:</p>

```python
s[2] = np.nan
```

```python
s.sort_values()
```

```python
s.sort_values(na_position="first")
```

<p>Sorting also supports a <code>key</code> parameter that takes a callable function to apply to the values being sorted.</p>

```python
s1 = pd.Series(["B", "a", "C"])

s1.sort_values()
```

```python
s1.sort_values(key=lambda x: x.str.lower())
```

<p><code>key</code> will be given the <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> of values and should return a <code>Series</code> or array of the same shape with the transformed values. For <code>DataFrame</code> objects,
the key is applied per column, so the key should still expect a Series and return a Series, e.g.</p>

```python
df = pd.DataFrame({"a": ["B", "a", "C"], "b": [1, 2, 3]})

df.sort_values(by="a")
```

```python
df.sort_values(by="a", key=lambda col: col.str.lower())
```

<p>The name or type of each column can be used to apply different functions to different columns.</p>

## <h3>By indexes and values</h3>

<p>Strings passed as the <code>by</code> parameter to <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html#pandas.DataFrame.sort_values" title="pandas.DataFrame.sort_values"><code>DataFrame.sort_values()</code></a> may
refer to either columns or index level names.</p>

```python
# Build MultiIndex
idx = pd.MultiIndex.from_tuples(
       [("a", 1), ("a", 2), ("a", 2), ("b", 2), ("b", 1), ("b", 1)]
   )

idx.names = ["first", "second"]
```

```python
# Build DataFrame
df_multi = pd.DataFrame({"A": np.arange(6, 0, -1)}, index=idx)
```

```python
df_multi
```

<p>Sort by ‘second’ (index) and ‘A’ (column)</p>

```python
df_multi.sort_values(by=["second", "A"])
```

<div class="alert alert-block alert-info">
<p>Note</p>
<p>If a string matches both a column name and an index level name then a
warning is issued and the column takes precedence. This will result in an ambiguity error in a future version.</p>
</div>

## <h3>smallest / largest values</h3>

<p><code>Series</code> has the <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.nsmallest.html" title="pandas.Series.nsmallest"><code>nsmallest()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.nlargest.html" title="pandas.Series.nlargest"><code>nlargest()</code></a> methods which return the smallest or largest n values. For a large <code>Series</code> this can be much faster than sorting the entire Series and calling <code>head(n)</code> on the result.</p>

```python
s = pd.Series(np.random.permutation(10))
```

```python
s
```

```python
s.sort_values()
```

```python
s.nsmallest(3)
```

```python
s.nlargest(3)
```

<p><code>DataFrame</code> also has the <code>nlargest</code> and <code>nsmallest</code> methods.</p>

```python
df.nlargest(3, "a")
```

```python
df.nsmallest(5, ["a", "c"])
```

# <h2>Copying</h2>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.copy.html" title="pandas.DataFrame.copy"><code>copy()</code></a> method on pandas objects copies the underlying data (though not the axis indexes, since they are immutable) and returns a new object.
Note that <strong>it is seldom necessary to copy objects</strong>. For example, there are only a handful of ways to alter a DataFrame <em>in-place</em>:</p>

<ul>
<li><p>Inserting, deleting, or modifying a column.</p></li>
<li><p>Assigning to the <code>index</code> or <code>columns</code> attributes.</p></li>
<li><p>For homogeneous data, directly modifying the values via the <code>values</code> attribute or advanced indexing.</p></li>
</ul>

<p>To be clear, no pandas method has the side effect of modifying your data; almost every method returns a new object, leaving the original object untouched. If the data is modified, it is because you did so explicitly.</p>

<p>To force a conversion, we can pass in an <code>errors</code> argument, which specifies how pandas should deal with elements that cannot be converted to desired dtype or object. By default, <code>errors='raise'</code>, meaning that any errors encountered

```python
m = ["apple", pd.Timedelta("1day")]

pd.to_timedelta(m, errors="coerce")
```

# <h2>Selecting columns based on <code>dtype</code></h2>

<p id="basics-selectdtypes">The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.select_dtypes.html#pandas.DataFrame.select_dtypes" title="pandas.DataFrame.select_dtypes"><code>select_dtypes()</code></a> method implements subsetting of columns based on their <code>dtype</code>.</p>

<p>First, let’s create a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> with a slew of different dtypes:</p>

```python
df = pd.DataFrame(
       {
           "string": list("abc"),
           "int64": list(range(1, 4)),
           "uint8": np.arange(3, 6).astype("u1"),
           "float64": np.arange(4.0, 7.0),
           "bool1": [True, False, True],
           "bool2": [False, True, False],
           "dates": pd.date_range("now", periods=3),
           "category": pd.Series(list("ABC")).astype("category"),
       }
   )
```

```python
df["tdeltas"] = df.dates.diff()
```

```python
df["uint64"] = np.arange(3, 6).astype("u8")
```

```python
df["other_dates"] = pd.date_range("20130101", periods=3)
```

```python
df["tz_aware_dates"] = pd.date_range("20130101", periods=3, tz="US/Eastern")
```

```python
df
```

<p>And the dtypes:</p>

```python
df.dtypes
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.select_dtypes.html#pandas.DataFrame.select_dtypes" title="pandas.DataFrame.select_dtypes"><code>select_dtypes()</code></a> has two parameters <code>include</code> and <code>exclude</code> that allow you to say “give me the columns <em>with</em> these dtypes” (<code>include</code>) and/or “give the columns <em>without</em> these dtypes” (<code>exclude</code>).</p>

<p>For example, to select <code>bool</code> columns:</p>

```python
df.select_dtypes(include=[bool])
```

<p>You can also pass the name of a dtype in the <a href="https://numpy.org/doc/stable/reference/arrays.scalars.html">NumPy dtype hierarchy</a>:</p>

```python
df.select_dtypes(include=["bool"])
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.select_dtypes.html#pandas.DataFrame.select_dtypes" title="pandas.DataFrame.select_dtypes"><code>select_dtypes()</code></a> also works with generic dtypes as well.</p>

<p>For example, to select all numeric and boolean columns while excluding unsigned integers:</p>

```python
df.select_dtypes(include=["number", "bool"], exclude=["unsignedinteger"])
```

<p>To select string columns you must use the <code>object</code> dtype:</p>

```python
df.select_dtypes(include=["object"])
```

<p>To see all the child dtypes of a generic <code>dtype</code> like <code>numpy.number</code> you can define a function that returns a tree of child dtypes:</p>

```python
def subdtypes(dtype):
    subs = dtype.__subclasses__()
    if not subs:
        return dtype
    return [dtype, [subdtypes(dt) for dt in subs]]
```

<p>All NumPy dtypes are subclasses of <code>numpy.generic</code>:</p>

```python
subdtypes(np.generic)
```
