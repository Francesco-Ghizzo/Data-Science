<h1>Merge, join, concatenate and compare</h1>

<p>pandas provides various methods for combining and comparing <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.</p>

<ul class="simple">

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.concat.html#pandas.concat" title="pandas.concat"><code>concat()</code></a>: Merge multiple <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects along a shared index or column</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html#pandas.DataFrame.join" title="pandas.DataFrame.join"><code>DataFrame.join()</code></a>: Merge multiple <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects along the columns</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.combine_first.html#pandas.DataFrame.combine_first" title="pandas.DataFrame.combine_first"><code>DataFrame.combine_first()</code></a>: Update missing values with non-missing values in the same location</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.merge.html#pandas.merge" title="pandas.merge"><code>merge()</code></a>: Combine two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects with SQL-style joining</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.merge_ordered.html#pandas.merge_ordered" title="pandas.merge_ordered"><code>merge_ordered()</code></a>: Combine two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects along an ordered axis</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.merge_asof.html#pandas.merge_asof" title="pandas.merge_asof"><code>merge_asof()</code></a>: Combine two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects by near instead of exact matching keys</p></li>

<li><p><a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.compare.html#pandas.Series.compare" title="pandas.Series.compare"><code>Series.compare()</code></a> and <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.compare.html#pandas.DataFrame.compare" title="pandas.DataFrame.compare"><code>DataFrame.compare()</code></a>: Show differences in values between two <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects</p></li>

</ul>

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.concat.html#pandas.concat" title="pandas.concat"><code>concat()</code></a></h2>

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.concat.html#pandas.concat" title="pandas.concat"><code>concat()</code></a> function concatenates an arbitrary amount of <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> or <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects along an axis while performing optional set logic (union or intersection) of the indexes on the other axes.
Like <code>numpy.concatenate</code>, <a href="https://pandas.pydata.org/docs/reference/api/pandas.concat.html#pandas.concat" title="pandas.concat"><code>concat()</code></a> takes a list or dict of homogeneously-typed objects and concatenates them.</p>

```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A2"],
        "B": ["B0", "B1", "B2", "B3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    },
    index=[0, 1, 2, 3],
)
```

```python
df2 = pd.DataFrame(
    {
        "A": ["A4", "A5", "A6", "A7"],
        "B": ["B4", "B5", "B6", "B7"],
        "C": ["C4", "C5", "C6", "C7"],
        "D": ["D4", "D5", "D6", "D7"],
    },
    index=[4, 5, 6, 7],
)
```

```python
df3 = pd.DataFrame(
    {
        "A": ["A8", "A9", "A10", "A11"],
        "B": ["B8", "B9", "B10", "B11"],
        "C": ["C8", "C9", "C10", "C11"],
        "D": ["D8", "D9", "D10", "D11"],
    },
    index=[8, 9, 10, 11],
)
```

```python
frames = [df1, df2, df3]
```

```python
result = pd.concat(frames)
```

```python
result
```

## <h3>Joining logic of the resulting axis</h3>

<p>The <code>join</code> keyword specifies how to handle axis values that don’t exist in the first <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.</p>

<p><code>join='outer'</code> takes the union of all axis values</p>

```python
df4 = pd.DataFrame(
    {
        "B": ["B2", "B3", "B6", "B7"],
        "D": ["D2", "D3", "D6", "D7"],
        "F": ["F2", "F3", "F6", "F7"],
    },
    index=[2, 3, 6, 7],
)
```

```python
result = pd.concat([df1, df4], axis=1)
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_concat_axis1.png">

<p><code>join='inner'</code> takes the intersection of the axis values</p>

```python
result = pd.concat([df1, df4], axis=1, join="inner")
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_concat_axis1_inner.png">

<p>To perform an effective “left” join using the <em>exact index</em> from the original <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>, result can be reindexed.</p>

```python
result = pd.concat([df1, df4], axis=1).reindex(df1.index)
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_concat_axis1_join_axes.png">

<p>For <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects which don’t have a meaningful index, the <code>ignore_index</code>
ignores overlapping indexes.</p>

```python
result
```

# <h2><a href="../reference/api/pandas.merge.html#pandas.merge" title="pandas.merge"><code>merge()</code></a></h2>

<p><a href="../reference/api/pandas.merge.html#pandas.merge" title="pandas.merge"><code>merge()</code></a> performs join operations similar to relational databases like SQL.
Users who are familiar with SQL but new to pandas can reference a
<a href="../getting_started/comparison/comparison_with_sql.html#compare-with-sql-join">comparison with SQL</a>.</p>

## <h3>Merge types</h3>

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.merge.html#pandas.merge" title="pandas.merge"><code>merge()</code></a> implements common SQL style joining operations.</p>

<ul class="simple">

<li><p><strong>one-to-one</strong>: joining two <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects on
their indexes which must contain unique values.</p></li>

<li><p><strong>many-to-one</strong>: joining a unique index to one or
more columns in a different <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.</p></li>

<li><p><strong>many-to-many</strong> : joining columns on columns.</p></li>

</ul>

<div class="alert alert-block alert-info">
<p>Note</p>
<p>When joining columns on columns, potentially a many-to-many join, any
indexes on the passed <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> objects <strong>will be discarded</strong>.</p>
</div>

<p>For a <strong>many-to-many</strong> join, if a key combination appears more than once in both tables, the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> will have the <strong>Cartesian product</strong> of the associated data.</p>

```python
left = pd.DataFrame(
    {
        "key": ["K0", "K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)
```

```python
right = pd.DataFrame(
    {
        "key": ["K0", "K1", "K2", "K3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    }
)
```

```python
result = pd.merge(left, right, on="key")
```

```python
result
```

If a key combination <strong>does not appear</strong> in
either the left or right tables, the values in the joined table will be
<code>NA</code>. Here is a summary of the <code>how</code> options and their SQL equivalent names:</p>

```python
left = pd.DataFrame(
   {
      "key1": ["K0", "K0", "K1", "K2"],
      "key2": ["K0", "K1", "K0", "K1"],
      "A": ["A0", "A1", "A2", "A3"],
      "B": ["B0", "B1", "B2", "B3"],
   }
)
```

```python
right = pd.DataFrame(
   {
      "key1": ["K0", "K1", "K1", "K2"],
      "key2": ["K0", "K0", "K0", "K0"],
      "C": ["C0", "C1", "C2", "C3"],
      "D": ["D0", "D1", "D2", "D3"],
   }
)
```

```python
result = pd.merge(left, right, how="left", on=["key1", "key2"])
```

<p>You can <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> and a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> with a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html#pandas.MultiIndex" title="pandas.MultiIndex"><code>MultiIndex</code></a> if the names of
the <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html#pandas.MultiIndex" title="pandas.MultiIndex"><code>MultiIndex</code></a> correspond to the columns from the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.
Transform the <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.html#pandas.Series" title="pandas.Series"><code>Series</code></a> to a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> using <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.reset_index.html#pandas.Series.reset_index" title="pandas.Series.reset_index"><code>Series.reset_index()</code></a> before merging</p>

```python
df = pd.DataFrame({"Let": ["A", "B", "C"], "Num": [1, 2, 3]})
```

```python
df
```

```python
ser = pd.Series(
    ["a", "b", "c", "d", "e", "f"],
    index=pd.MultiIndex.from_arrays(
        [["A", "B", "C"] * 2, [1, 2, 3, 4, 5, 6]], names=["Let", "Num"]
    ),
)
```

```python
ser
```

```python
pd.merge(df, ser.reset_index(), on=["Let", "Num"])
```

```python
right = pd.DataFrame({"A": [4, 5, 6], "B": [2, 2, 2]})
```

```python
result = pd.merge(left, right, on="B", how="outer")
```

## <h3>Merge key uniqueness</h3>

<p>The <code>validate</code> argument checks whether the uniqueness of merge keys.
Key uniqueness is checked before merge operations and can protect against memory overflows and unexpected key duplication.</p>

```python
left = pd.DataFrame({"A": [1, 2], "B": [1, 2]})
```

```python
right = pd.DataFrame({"A": [4, 5, 6], "B": [2, 2, 2]})
```

```python
result = pd.merge(left, right, on="B", how="outer", validate="one_to_one")
```

<p>If the user is aware of the duplicates in the right <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> but wants to ensure there are no duplicates in the left <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>, one can use the <code>validate='one_to_many'</code> argument instead, which will not raise an exception.</p>

```python
pd.merge(left, right, on="B", how="outer", validate="one_to_many")
```

# <h2><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html#pandas.DataFrame.join" title="pandas.DataFrame.join"><code>DataFrame.join()</code></a></h2>

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html#pandas.DataFrame.join" title="pandas.DataFrame.join"><code>DataFrame.join()</code></a> combines the columns of multiple, potentially differently-indexed <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> into a single result
<a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.</p>

```python
right = pd.DataFrame(
    {"C": ["C0", "C2", "C3"], "D": ["D0", "D2", "D3"]}, index=["K0", "K2", "K3"]
)
```

```python
result = left.join(right)
```

```python
result
```

```python
result
```

<p><a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html#pandas.DataFrame.join" title="pandas.DataFrame.join"><code>DataFrame.join()</code></a> takes an optional <code>on</code> argument which may be a column or multiple column names that the passed <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> is to be aligned.</p>

```python
left = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
        "key": ["K0", "K1", "K0", "K1"],
    }
)
```

```python
right = pd.DataFrame({"C": ["C0", "C1"], "D": ["D0", "D1"]}, index=["K0", "K1"])
```

```python
result = left.join(right, on="key")
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_join_key_columns.png">

```python
result = pd.merge(
    left, right, left_on="key", right_index=True, how="left", sort=False
)
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_merge_key_columns.png">
<p id="merging-multikey-join">To join on multiple keys, the passed <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> must have a <a href="https://pandas.pydata.org/docs/reference/api/pandas.MultiIndex.html#pandas.MultiIndex" title="pandas.MultiIndex"><code>MultiIndex</code></a>:</p>

```python
left = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
        "key1": ["K0", "K0", "K1", "K2"],
        "key2": ["K0", "K1", "K0", "K1"],
    }
)
```

```python
index = pd.MultiIndex.from_tuples(
    [("K0", "K0"), ("K1", "K0"), ("K2", "K0"), ("K2", "K1")]
)
```

```python
right = pd.DataFrame(
    {"C": ["C0", "C1", "C2", "C3"], "D": ["D0", "D1", "D2", "D3"]}, index=index
)
```

```python
result = left.join(right, on=["key1", "key2"])
```

```python
result
```

<img src="https://pandas.pydata.org/docs/_images/merging_join_multikeys.png">

<p id="merging-df-inner-join">The default for <a href="../reference/api/pandas.DataFrame.join.html#pandas.DataFrame.join" title="pandas.DataFrame.join"><code>DataFrame.join</span></code></a> is to perform a left join which uses only the keys found in the calling <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</span></code></a>. Other join types can be specified with <code>how</span></code>.</p>

```python
result = left.join(right, on=["key1", "key2"], how="inner")
```

For each row in the <code>left</code> <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>, the last row in the <code>right</code> <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> are selected where the <code>on</code> key is less than the left’s key.
Both <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a> must be sorted by the key.</p>

<p>Optionally an <a href="https://pandas.pydata.org/docs/reference/api/pandas.merge_asof.html#pandas.merge_asof" title="pandas.merge_asof"><code>merge_asof()</code></a> can perform a group-wise merge by matching the
<code>by</code> key in addition to the nearest match on the <code>on</code> key.</p>
