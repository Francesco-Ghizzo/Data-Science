<h1>Windowing operations</h1>

<p>pandas contains a compact set of APIs for performing windowing operations - an operation that performs an aggregation over a sliding partition of values.
The API functions similarly to the <code>groupby</code> API in that
<a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> and
<a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> call the windowing method with necessary parameters and then subsequently call the aggregation function.</p>

```python
import pandas as pd
```

```python
s = pd.Series(range(5))
```

```python
s.rolling(window=2).sum()
```

<p>The windows are comprised by looking back the length of the window from the current observation.
The result above can be derived by taking the sum of the following windowed partitions of data:</p>

```python
for window in s.rolling(window=2):
    print(window)
```

```python
s = pd.Series(range(5), index=pd.date_range('2020-01-01', periods=5, freq='1D'))
```

```python
s.rolling(window='2D').sum()
```

<p>Additionally, some methods support chaining a <code>groupby</code> operation with a windowing operation which will first group the data by the specified keys and then perform a windowing operation per group.</p>

```python
df = pd.DataFrame({'A': ['a', 'b', 'a', 'b', 'a'], 'B': range(5)})
```

```python
df.groupby('A').expanding().sum()
```

<p>Some windowing operations also support the <code>method='table'</code> option in the constructor which performs the windowing operation over an entire <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> instead of a single column or row at a time.
This can provide a useful performance benefit for a <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> with many columns or rows (with the corresponding <code>axis</code> argument) or the ability to utilize other columns during the windowing operation. The <code>method='table'</code> option can only be used if <code>engine='numba'</code> is specified in the corresponding method call.</p>

<p>For example, a <a href="https://en.wikipedia.org/wiki/Weighted_arithmetic_mean">weighted mean</a> calculation can be calculated with <code>apply()</code> by specifying a separate column of weights.</p>

```python
import numpy as np
```

```python
def weighted_mean(x):
    arr = np.ones((1, x.shape[1]))
    arr[:, :2] = (x[:, :2] * x[:, 2]).sum(axis=0) / x[:, 2].sum()
    return arr
```

```python
df = pd.DataFrame([[1, 2, 0.6], [2, 3, 0.4], [3, 4, 0.2], [4, 5, 0.7]])
```

```python
df.rolling(2, method="table", min_periods=0).apply(weighted_mean, raw=True, engine="numba")  # noqa: E501
```

<div class="versionadded">
<p>Added in version 1.3.</span></p>
</div>

<p>Some windowing operations also support an <code>online</span></code> method after constructing a windowing object which returns a new object that supports passing in new <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> or <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> objects to continue the windowing calculation with the new values (i.e. online calculations).</p>

<p>The methods on this new windowing objects must call the aggregation method first to “prime” the initial state of the online calculation.
Then, new <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> or <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="pandas.Series"><code>Series</code></a> objects can be passed in the <code>update</code> argument to continue the windowing calculation.</p>

```python
df = pd.DataFrame([[1, 2, 0.6], [2, 3, 0.4], [3, 4, 0.2], [4, 5, 0.7]])
```

```python
df.ewm(0.5).mean()
```

```python
online_ewm = df.head(2).ewm(0.5).online()
```

```python
online_ewm.mean()
```

```python
online_ewm.mean(update=df.tail(1))
```

<p>All windowing operations support a <code>min_periods</code> argument that dictates the minimum amount of non-<code>np.nan</code> values a window must have; otherwise, the resulting value is <code>np.nan</span></code>.
<code>min_periods</span></code> defaults to 1 for time-based windows and <code>window</span></code> for fixed windows</p>

```python
s = pd.Series([np.nan, 1, 2, np.nan, np.nan, 3])
```

```python
s.rolling(window=3, min_periods=1).sum()
```

```python
s.rolling(window=3, min_periods=2).sum()
```

```python
# Equivalent to min_periods=3
s.rolling(window=3, min_periods=None).sum()
```

<p>Additionally, all windowing operations supports the <code>aggregate</code> method for returning a result of multiple aggregations applied to a window.</p>

```python
df = pd.DataFrame({"A": range(5), "B": range(10, 15)})
```

```python
df.expanding().agg(["sum", "mean", "std"])
```

# <h2>Rolling window</h2>

<p>Generic rolling windows support specifying windows as a fixed number of observations or variable number of observations based on an offset.
If a time based offset is provided, the corresponding time based index must be monotonic.</p>

```python
times = ['2020-01-01', '2020-01-03', '2020-01-04', '2020-01-05', '2020-01-29']
```

```python
s = pd.Series(range(5), index=pd.DatetimeIndex(times))
```

```python
s
```

```python
# Window with 2 observations
s.rolling(window=2).sum()
```

```python
# Window with 2 days worth of observations
s.rolling(window='2D').sum()
```

<p>For all supported aggregation functions, see <a href="https://pandas.pydata.org/docs/reference/window.html#api-functions-rolling">Rolling window functions</a>.</p>

## <h3>Centering windows</h3>

<p>By default the labels are set to the right edge of the window, but a <code>center</code> keyword is available so the labels can be set at the center.</p>

```python
s = pd.Series(range(10))
```

```python
s.rolling(window=5).mean()
```

```python
s.rolling(window=5, center=True).mean()
```

<p>This can also be applied to datetime-like indices.</p>

<div class="versionadded">
<p>Added in version 1.3.0.</p>
</div>

```python
df = pd.DataFrame(
    {"A": [0, 1, 2, 3, 4]}, index=pd.date_range("2020", periods=5, freq="1D")
)
```

```python
df
```

```python
df.rolling("2D", center=False).mean()
```

```python
df.rolling("2D", center=True).mean()
```

## <h3>Numba engine</h3>

<p>Additionally, <code>apply()</span></code> can leverage <a href="https://numba.pydata.org/">Numba</a> if installed as an optional dependency.
The apply aggregation can be executed using Numba by specifying <code>engine='numba'</span></code> and <code>engine_kwargs</code> arguments (<code>raw</code> must also be set to <code>True</code>).
See <a href="enhancingperf.html#enhancingperf-numba">enhancing performance with Numba</a> for general usage of the arguments and performance considerations.</p>

<p>Numba will be applied in potentially two routines:</p>

<ol class="arabic simple">
<li><p>If <code>func</code> is a standard Python function, the engine will <a href="https://numba.pydata.org/numba-doc/latest/user/overview.html">JIT</a> the passed function. <code>func</span></code> can also be a JITed function in which case the engine will not JIT the function again.</p></li>
<li><p>The engine will JIT the for loop where the apply function is applied to each window.</p></li>
</ol>

<p>The <code>engine_kwargs</code> argument is a dictionary of keyword arguments that will be passed into the
<a href="https://numba.pydata.org/numba-doc/latest/reference/jit-compilation.html#numba.jit">numba.jit decorator</a>.
These keyword arguments will be applied to <em>both</em> the passed function (if a standard Python function) and the apply for loop over each window.</p>
                                                                                                                           
<div class="versionadded">
<p>Added in version 1.3.0.</p>
</div>

<p><code>mean</code>, <code>median</code>, <code>max</code>, <code>min</code>, and <code>sum</code> also support the <code>engine</code> and <code>engine_kwargs</code> arguments.</p>

## <h3>Computing rolling pairwise covariances and correlations</h3>

<p>In financial data analysis and other fields it’s common to compute covariance and correlation matrices for a collection of time series. Often one is also interested in moving-window covariance and correlation matrices. This can be done by passing the <code>pairwise</span></code> keyword argument, which in the case of <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</code></a> inputs will yield a MultiIndexed <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="pandas.DataFrame"><code>DataFrame</span></code></a> whose <code>index</span></code> are the dates in question. In the case of a single DataFrame argument the <code>pairwise</span></code> argument can even be omitted:</p>

<div class="admonition note">
<p class="admonition-title">Note</p>
<p>Missing values are ignored and each entry is computed using the pairwise complete observations.</p>

<p>Assuming the missing data are missing at random this results in an estimate for the covariance matrix which is unbiased. However, for many applications this estimate may not be acceptable because the estimated covariance matrix is not guaranteed to be positive semi-definite.

See <a href="https://en.wikipedia.org/w/index.php?title=Estimation_of_covariance_matrices">Estimation of covariance matrices</a> for more details.</p>
