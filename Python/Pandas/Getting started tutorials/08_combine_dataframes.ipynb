<h1>How to combine data from multiple tables</h1>

# <h2>Concatenating objects</h2>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/08_concat_row.svg">

### <h4>I want to combine the measurements of $NO_2$ and $PM_{25}$, two tables with a similar structure, in a single table.</h4>

```python
air_quality = pd.concat([air_quality_pm25, air_quality_no2], axis=0)
```

```python
air_quality.head()
```

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.concat.html#pandas.concat" title="pandas.concat"><code>concat()</code></a> function performs concatenation operations of multiple
tables along one of the axes (row-wise or column-wise).</p>

<p>By default concatenation is along axis 0, so the resulting table combines the rows of the input tables. Let’s check the shape of the original and the concatenated tables to verify the operation:</p>

```python
print('Shape of the air_quality_pm25 table: ', air_quality_pm25.shape)
```

```python
print('Shape of the air_quality_no2 table: ', air_quality_no2.shape)
```

```python
print('Shape of the resulting air_quality table: ', air_quality.shape)
```

<p>Hence, the resulting table has 3178 = 1110 + 2068 rows.</p>

<div class="alert alert-block alert-info">
<p>Note</p>
<p>The <strong>axis</strong> argument will return in a number of pandas
methods that can be applied <strong>along an axis</strong>.
A <code>DataFrame</code> has two corresponding axes: the first running vertically downwards across rows (axis 0), and the second running horizontally across columns (axis 1).
Most operations like concatenation or summary statistics are by default
across rows (axis 0), but can be applied across columns as well.</p>
</div>

<p>Sorting the table on the datetime information illustrates also the
combination of both tables, with the <code>parameter</code> column defining the origin of the table (either <code>no2</code> from table <code>air_quality_no2</code> or <code>pm25</code> from table <code>air_quality_pm25</code>):</p>

```python
air_quality = air_quality.sort_values("date.utc")
```

```python
air_quality.head()
```

<p>In this specific example, the <code>parameter</code> column provided by the data ensures that each of the original tables can be identified. This is not always the case. The <code>concat</code> function provides a convenient solution with the <code>keys</code> argument, adding an additional (hierarchical) row index. For example:</p>

```python
air_quality_ = pd.concat([air_quality_pm25, air_quality_no2], keys=["PM25", "NO2"])
```

```python
air_quality_.head()
```

<div class="alert alert-block alert-info">
<p>Note</p>
<p>The existence of multiple row/column indices at the same time
has not been mentioned within these tutorials.
<em>Hierarchical indexing</em> or <em>MultiIndex</em> is an advanced and powerful pandas feature to analyze higher dimensional data.</p>
<p>Multi-indexing is out of scope for this pandas introduction.
For the moment, remember that the function <code>reset_index</code> can be used to convert any level of an index to a column, e.g.
<code>air_quality.reset_index(level=0)</code></p>
</div>

# <h2>Join tables using a common identifier</h2>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/08_merge_left.svg">

### <h4>Add the station coordinates, provided by the stations metadata table, to the corresponding rows in the measurements table.</h4>

<div class="alert alert-block alert-danger">
<p>Warning</p>
<p>The air quality measurement station coordinates are stored in a data
file <code>air_quality_stations.csv</code>, downloaded using the
<a href="http://dhhagan.github.io/py-openaq/index.html">py-openaq</a> package.</p>
</div>

```python
!curl https://raw.githubusercontent.com/pandas-dev/pandas/refs/heads/main/doc/data/air_quality_stations.csv -o air_quality_stations.csv
```

```python
stations_coord = pd.read_csv("air_quality_stations.csv")
```

```python
stations_coord.head()
```

<div class="alert alert-block alert-info">
<p>Note</p>
<p>The stations used in this example (FR04014, BETR801 and London
Westminster) are just three entries enlisted in the metadata table.
We only want to add the coordinates of these three to the measurements
table, each on the corresponding rows of the <code>air_quality</code> table.</p>
</div>

```python
air_quality.head()
```

```python
air_quality = pd.merge(air_quality, stations_coord, how="left", on="location")
```

```python
air_quality.head()
```

<p>Using the <a href="https://pandas.pydata.org/docs/reference/api/pandas.merge.html#pandas.merge" title="pandas.merge"><code>merge()</code></a> function, for each of the rows in the <code>air_quality</code> table, the corresponding coordinates are added from the <code>air_quality_stations_coord</code> table.
Both tables have the column <code>location</code> in common which is used as a key to combine the information.
By choosing the <code>left</code> join, only the locations available in the <code>air_quality</code> (left) table, i.e.&nbsp;FR04014, BETR801 and London Westminster, end up in the resulting table.
The <code>merge</code> function supports multiple join options similar to database-style operations.</p>

### <h4>Add the parameters’ full description and name, provided by the parameters metadata table, to the measurements table.</h4>

<div class="alert alert-block alert-danger">
<p>Warning</p>
<p>The air quality parameters metadata are stored in a data file
<code>air_quality_parameters.csv</code>, downloaded using the
<a href="http://dhhagan.github.io/py-openaq/index.html">py-openaq</a> package.</p>
</div>

```python
!curl https://raw.githubusercontent.com/pandas-dev/pandas/8cf2e43e1ca2a50f010fd37da5a01e8a3a6f1822/doc/data/air_quality_parameters.csv -o air_quality_parameters.csv
```

```python
air_quality_parameters = pd.read_csv("air_quality_parameters.csv")
```

```python
air_quality_parameters.head()
```

```python
air_quality = pd.merge(air_quality, air_quality_parameters, how='left', left_on='parameter', right_on='id')
```

```python
air_quality.head()
```

<p>Compared to the previous example, there is no common column name.
However, the <code>parameter</code> column in the <code>air_quality</code> table and the <code>id</code> column in the <code>air_quality_parameters_name</code> both provide the measured variable in a common format.
The <code>left_on</code> and <code>right_on</code> arguments are used here (instead of just <code>on</code>) to make the link between the two tables.</p>

<div>
<p><b>To user guide</b> Pandas supports also inner, outer, and right joins. More information on join/merge of tables is provided in the user guide section on <a href="https://pandas.pydata.org/docs/user_guide/merging.html#merging-join">database style merging of tables</a>.
Or have a look at the <a href="https://pandas.pydata.org/docs/getting_started/comparison/comparison_with_sql.html#compare-with-sql-join">comparison with SQL</a> page.</p>
