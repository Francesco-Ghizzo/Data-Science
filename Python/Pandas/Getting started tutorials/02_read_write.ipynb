<h1>How do I read and write tabular data?</h1>

### <h4>I want to analyze the Titanic passenger data, available as a CSV file.</h4>

```python
titanic = pd.read_csv("titanic.csv")
```

<p>pandas provides the <a href="https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html#pandas.read_csv" title="pandas.read_csv"><code>read_csv()</code></a> function to read data stored as a csv file into a pandas <code>DataFrame</code>. Pandas supports many different file formats or data sources out of the box (csv, excel, sql, json, parquet, …), each of them with the prefix <code>read_*</code>.</p>

<p>Make sure to always have a check on the data after reading in the data.
When displaying a <code>DataFrame</code>, the first and last 5 rows will be shown by default:</p>

```python
titanic
```

### <h4>I want to see the first 8 rows of a pandas DataFrame.</h4>

```python
titanic.head(8)
```

<p>To see the first N rows of a <code>DataFrame</code>, use the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html#pandas.DataFrame.head" title="pandas.DataFrame.head"><code>head()</code></a> method with
the required number of rows (in this case 8) as argument.</p>
    
<div class="alert alert-block alert-info">
<p>Note</p>
<p>Interested in the last N rows instead? pandas also provides a
<a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.tail.html#pandas.DataFrame.tail" title="pandas.DataFrame.tail"><code>tail()</code></a> method. For example, <code>titanic.tail(10)</code> will return the last 10 rows of the DataFrame.</p>
</div>

<p>A check on how pandas interpreted each of the column data types can be
done by requesting the pandas <code>dtypes</code> attribute:</p>

```python
titanic.dtypes
```

<p>For each of the columns, the used data type is enlisted. The data types
in this <code>DataFrame</code> are integers (<code>int64</code>), floats (<code>float64</code>) and strings (<code>object</code>).</p>

<div class="alert alert-block alert-info">
<p>Note</p>
<p>When asking for the <code>dtypes</code>, no brackets are used!
<code>dtypes</code> is an attribute of a <code>DataFrame</code> and <code>Series</code>. Attributes of a <code>DataFrame</code> or <code>Series</code> do not need brackets. Attributes represent a characteristic of a <code>DataFrame</code>/<code>Series</code>, whereas methods (which require brackets) <em>do</em> something with the
<code>DataFrame</code>/<code>Series</code> as introduced in the <a href="https://pandas.pydata.org/docs/getting_started/intro_tutorials/01_table_oriented.html#min-tut-01-tableoriented">first tutorial</a>.</p>
</div>

### <h4>My colleague requested the Titanic data as a spreadsheet.</h4>

```python
titanic.to_excel("titanic.xlsx", sheet_name="passengers", index=False)
```

<p>Whereas <code>read_*</code> functions are used to read data to pandas, the <code>to_*</code> methods are used to store data. The <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.to_excel.html#pandas.DataFrame.to_excel" title="pandas.DataFrame.to_excel"><code>to_excel()</code></a> method stores the data as an excel file.
In the example here, the <code>sheet_name</code> is named <em>passengers</em> instead of the default <em>Sheet1</em>.
By setting <code>index=False</code> the row index labels are not saved in the spreadsheet.</p>

<p>The equivalent read function <code>read_excel()</code> will reload the data to a <code>DataFrame</code>:</p>

```python
titanic = pd.read_excel("titanic.xlsx", sheet_name="passengers")
```

```python
titanic.head()
```

### <h4>I’m interested in a technical summary of a <code>DataFrame</code></h4>

```python
titanic.info()
```

<p>The method <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.info.html#pandas.DataFrame.info" title="pandas.DataFrame.info"><code>info()</code></a> provides technical information about a
<code>DataFrame</code>, so let’s explain the output in more detail:</p>

<ul>

<li><p>It is indeed a <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html#pandas.DataFrame" title="pandas.DataFrame"><code>DataFrame</code></a>.</p></li>

<li><p>There are 891 entries, i.e.&nbsp;891 rows.</p></li>

<li><p>Each row has a row label (aka the <code>index</code>) with values ranging from 0 to 890.</p></li>

<li><p>The table has 12 columns. Most columns have a value for each of the
rows (all 891 values are <code>non-null</code>). Some columns do have missing values and less than 891 <code>non-null</code> values.</p></li>

<li><p>The columns <code>Name</code>, <code>Sex</code>, <code>Cabin</code> and <code>Embarked</code> consists of textual data (strings, aka <code>object</code>). The other columns are numerical data with some of them whole numbers (aka <code>integer</code>) and others are real numbers (aka <code>float</code>).</p></li>

<li><p>The kind of data (characters, integers,…) in the different columns
are summarized by listing the <code>dtypes</code>.</p></li>

<li><p>The approximate amount of RAM used to hold the DataFrame is provided
as well.</p></li>

</ul>
