<h1>How to calculate summary statistics</h1>

# <h2>Aggregating statistics</h2>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/06_aggregate.svg">

### <h4>What is the average age of the Titanic passengers?</h4>

```python
titanic["Age"].mean()
```

<p>Different statistics are available and can be applied to columns with numerical data. Operations in general exclude missing data and operate across rows by default.</p>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/06_reduction.svg">

### <h4>What is the median age and ticket fare price of the Titanic passengers?</h4>

```python
titanic[["Age", "Fare"]].median()
```

<p>The statistic applied to multiple columns of a <code>DataFrame</code> (the selection of two columns returns a <code>DataFrame</code>, see the <a href="https://pandas.pydata.org/docs/getting_started/intro_tutorials/03_subset_data.html#min-tut-03-subset">subset data tutorial</a>) is calculated for each numeric column.</p>

<p>The aggregating statistic can be calculated for multiple columns at the same time. Remember the <code>describe</code> function from the <a href="https://pandas.pydata.org/docs/getting_started/intro_tutorials/01_table_oriented.html#min-tut-01-tableoriented">first tutorial</a>?</p>

```python
titanic[["Age", "Fare"]].describe()
```

<p>Instead of the predefined statistics, specific combinations of
aggregating statistics for given columns can be defined using the
<a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.agg.html#pandas.DataFrame.agg"><code>DataFrame.agg()</code></a> method:</p>

```python
titanic.agg(
    {
        "Age": ["min", "max", "median", "skew"],
        "Fare": ["min", "max", "median", "mean"],
    }
)
```

<div>
<p><b>To user guide</b> Details about descriptive statistics are provided in the user guide section on <a href="https://pandas.pydata.org/docs/user_guide/basics.html#basics-stats">descriptive statistics</a>.</p>
</div>

# <h2>Aggregating statistics grouped by category</h2>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/06_groupby.svg">

### <h4>What is the average age for male versus female Titanic passengers?</h4>


```python
titanic[["Sex", "Age"]].groupby("Sex").mean()
```

<p>As our interest is the average age for each gender, a subselection on
these two columns is made first: <code>titanic[["Sex", "Age"]]</code>. Next, the
<a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html#pandas.DataFrame.groupby" title="pandas.DataFrame.groupby"><code>groupby()</code></a> method is applied on the <code>Sex</code> column to make a group per category. The average age <em>for each gender</em> is calculated and returned.</p>
    
<p>Calculating a given statistic (e.g. <code>mean</code> age) <em>for each category in a column</em> (e.g.&nbsp;male/female in the <code>Sex</code> column) is a common pattern.
The <code>groupby</code> method is used to support this type of operations. This fits in the more general <code>split-apply-combine</code> pattern:</p>

<ul>
<li><p><strong>Split</strong> the data into groups</p></li>
<li><p><strong>Apply</strong> a function to each group independently</p></li>
<li><p><strong>Combine</strong> the results into a data structure</p></li>
</ul>

<p>The apply and combine steps are typically done together in pandas.</p>

<p>In the previous example, we explicitly selected the 2 columns first. If not, the <code>mean</code> method is applied to each column containing numerical columns by passing <code>numeric_only=True</code>:</p>

```python
titanic.groupby("Sex").mean(numeric_only=True)
```

<p>It does not make much sense to get the average value of the <code>Pclass</code>.
If we are only interested in the average age for each gender, the
selection of columns (rectangular brackets <code>[]</code> as usual) is supported on the grouped data as well:</p>

```python
titanic.groupby("Sex")["Age"].mean()
```

<img class="align-center" src="https://pandas.pydata.org/docs/_images/06_groupby_select_detail.svg">

<div class="alert alert-block alert-info">
<p>Note</p>
<p>The <code>Pclass</code> column contains numerical data but actually represents 3 categories (or factors) with respectively the labels ‘1’, ‘2’ and ‘3’. Calculating statistics on these does not make much sense.
Therefore, pandas provides a <code>Categorical</code> data type to handle this type of data. More information is provided in the user guide <a href="https://pandas.pydata.org/docs/user_guide/categorical.html#categorical">Categorical data</a> section.</p>
</div>

### <h4>What is the mean ticket fare price for each of the sex and cabin class combinations?</h4>

```python
titanic.groupby(["Sex", "Pclass"])["Fare"].mean()
```

<p>Grouping can be done by multiple columns at the same time. Provide the column names as a list to the <a href="https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html#pandas.DataFrame.groupby" title="pandas.DataFrame.groupby"><code>groupby()</code></a> method.</p>

<p><b>To user guide</b> A full description on the split-apply-combine approach is provided in the user guide section on <a href="https://pandas.pydata.org/docs/user_guide/groupby.html#groupby">groupby operations</a>.</p>
</div>

# <h2>Count number of records by category</h2>

<img class="align-center" src="https://pandas.pydata.org/docs/_images/06_valuecounts.svg">

### <h4>What is the number of passengers in each of the cabin classes?</h4>

```python
titanic["Pclass"].value_counts()
```

<p>The <a href="https://pandas.pydata.org/docs/reference/api/pandas.Series.value_counts.html#pandas.Series.value_counts" title="pandas.Series.value_counts"><code>value_counts()</code></a> method counts the number of records for each
category in a column.</p>
    
<p>The function is a shortcut, as it is actually a groupby operation in combination with counting of the number of records
within each group:</p>

```python
titanic.groupby("Pclass")["Pclass"].count()
```

<div class="alert alert-block alert-info">
<p>Note</p>
<p>Both <code>size</code> and <code>count</code> can be used in combination with <code>groupby</code>. Whereas <code>size</code> includes <code>NaN</code> values and just provides the number of rows (size of the table), <code>count</code> excludes the missing values.
In the <code>value_counts</code> method, use the <code>dropna</code> argument to include or exclude the <code>NaN</code> values.</p>
</div>

<div>
<p><b>To user guide</b> The user guide has a dedicated section on <code>value_counts</code> , see the page on <a href="https://pandas.pydata.org/docs/user_guide/basics.html#basics-discretization">discretization</a>.</p>
