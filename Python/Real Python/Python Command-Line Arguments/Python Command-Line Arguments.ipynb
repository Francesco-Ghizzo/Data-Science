# <h2 id="the-command-line-interface">The Command-Line Interface<a class="headerlink" href="#the-command-line-interface"></a></h2>

<p>A <a href="https://en.wikipedia.org/wiki/Command-line_interface"><strong>command-line interface (CLI)</strong></a> provides a way for a user to interact with a program running in a text-based <a href="https://en.wikipedia.org/wiki/Shell_%28computing%29">shell</a> interpreter. Some examples of shell interpreters are <a href="https://en.wikipedia.org/wiki/Bash_%28Unix_shell%29">Bash</a> on Linux or
<a href="https://en.wikipedia.org/wiki/Cmd.exe">Command Prompt</a> on Windows. A command-line interface is enabled by the shell interpreter that exposes a <a href="https://en.wikipedia.org/wiki/Command-line_interface#Command_prompt">command prompt</a>. It can be characterized by the following elements:</p>

<ul>
<li>A <strong>command</strong> or program</li>
<li>Zero or more command line <strong>arguments</strong></li>
<li>An <strong>output</strong> representing the result of the command</li>
<li>Textual documentation referred to as <strong>usage</strong> or <strong>help</strong></li>
</ul>

<p>Not every command-line interface may provide all these elements, but this list isn’t exhaustive, either. The complexity of the command line ranges from the ability to pass a single argument, to numerous arguments and options, much like a <a href="https://en.wikipedia.org/wiki/Domain-specific_language">Domain Specific Language</a>. For example, some programs may launch web documentation from the command line or start an
<a href="https://docs.python.org/tutorial/interpreter.html#interactive-mode">interactive shell interpreter</a> like Python.</p>

<p>The two following examples with the Python command illustrates the description of a command-line interface:</p>

```python
! python -c "print('Real Python')"
```

<p>In this first example, the Python interpreter takes option <code>-c</code> for <strong>command</strong>, which says to execute the Python command-line arguments following the option <code>-c</code> as a Python program.</p>

<p>Another example shows how to invoke Python with <code>-h</code> to display the help:</p>

```python
! python -h
```

<p>Try this out in your <a href="https://realpython.com/terminal-commands/">terminal</a> to see the complete help documentation.</p>

# <h2 id="the-c-legacy">The C Legacy<a class="headerlink" href="#the-c-legacy"></a></h2>

<p>Python command-line arguments directly inherit from the <a href="https://realpython.com/build-python-c-extension-module/">C</a> programming language. As <a href="https://en.wikipedia.org/wiki/Guido_van_Rossum">Guido Van Rossum</a> wrote in
<a href="http://citeseerx.ist.psu.edu/viewdoc/summary?doi=10.1.1.47.4180">An Introduction to Python for Unix/C Programmers</a> in 1993, C had a strong influence on Python. Guido mentions the definitions of literals, identifiers, operators, and statements like <code>break</code>, <code>continue</code>, or <code>return</code>. The use of Python command-line arguments is also strongly influenced by the C language.</p>
                                                                                                                                      
<p>To illustrate the similarities, consider the following C program:</p>

<pre><code>
// main.c
#include <stdio.h>

int main(int argc, char *argv[]) {
    printf("Arguments count: %d\n", argc);
    for (int i = 0; i < argc; i++) {
        printf("Argument %6d: %s\n", i, argv[i]);
    }
    return 0;
}
</code></pre>

<p>Line 4 defines <a href="https://en.wikipedia.org/wiki/Entry_point#C_and_C++"><code>main()</code></a>, which is the entry point of a C program. Take good note of the parameters:</p>

<ol>
<li><strong><code>argc</code></strong> is an integer representing the number of arguments of the program.</li>
<li><strong><code>argv</code></strong> is an array of pointers to characters containing the name of the program in the first element of the array, followed by the arguments of the program, if any, in the remaining elements of the array.</li>
</ol>

<p>You can compile the code above on Linux with <code>gcc -o main main.c</code>, then execute with <code>./main</code> to obtain the following:</p>

```python
! gcc -o main main.c
```

```python
! ./main
```

<p>Unless explicitly expressed at the command line with the option <code>-o</code>, <a href="https://en.wikipedia.org/wiki/A.out"><code>a.out</code></a> is the default name of the executable generated by the <strong>gcc</strong> compiler. It stands for <strong>assembler output</strong> and is reminiscent of the executables that were generated on older UNIX systems. Observe that the name of the executable <code>./main</code> is the sole argument.</p>

<p>Let’s spice up this example by passing a few Python command-line arguments to the same program:</p>

```python
! ./main Python Command Line Arguments
```

<p>The output shows that the number of arguments is <code>5</code>, and the list of arguments includes the name of the program, <code>main</code>, followed by each word of the phrase <code>"Python Command Line Arguments"</code>, which you passed at the command line.</p>

<div class="alert alert-primary" role="alert">
<p><strong>Note</strong>: <code>argc</code> stands for <strong>argument count</strong>, while <code>argv</code> stands for <strong>argument vector</strong>. To learn more, check out <a href="https://en.wikibooks.org/wiki/A_Little_C_Primer/C_Command_Line_Arguments">A Little C Primer/C Command Line Arguments</a>.</p>
</div>

<p>You could implement a Python program, <code>main.py</code>, that’s equivalent to the C program, <code>main.c</code>, you saw above:</p>

```python
# main.py
import sys

if __name__ == "__main__":
    print(f"Arguments count: {len(sys.argv)}")
    for i, arg in enumerate(sys.argv):
        print(f"Argument {i:>6}: {arg}")
```

<p>You don’t see an <code>argc</code> <a href="https://realpython.com/python-variables/">variable</a> like in the C code example. It doesn’t exist in Python because <code>sys.argv</code> is sufficient. You can parse the Python command-line arguments in <code>sys.argv</code> without having to know the length of the list, and you can call the built-in <a href="https://realpython.com/len-python-function/"><code>len()</code></a> if the number of arguments is needed by your program.</p>

<p>Also, note that <a href="https://realpython.com/python-enumerate/"><code>enumerate()</code></a>, when applied to an iterable, returns an <code>enumerate</code> object that can emit pairs associating the index of an element in <code>sys.arg</code> to its corresponding value. This allows looping through the content of <code>sys.argv</code> without having to maintain a counter for the index in the list.</p>

<p>Execute <code>main.py</code> as follows:</p>

```python
! python main.py Python Command Line Arguments
```

<p><code>sys.argv</code> contains the same information as in the C program:</p>

<ul>
<li><strong>The name of the program</strong> <code>main.py</code> is the first item of the list.</li>
<li><strong>The arguments</strong> <code>Python</code>, <code>Command</code>, <code>Line</code>, and <code>Arguments</code> are the remaining elements in the list.</li>
</ul>

<p>With this short introduction into a few arcane aspects of the C language, you’re now armed with some valuable knowledge to further grasp Python command-line arguments.</p>

## <h3 id="sha1sum"><code>sha1sum</code><a class="headerlink" href="#sha1sum"></a></h3>

<p><code>sha1sum</code> calculates <a href="https://en.wikipedia.org/wiki/SHA-1">SHA-1</a>
<a href="https://en.wikipedia.org/wiki/Cryptographic_hash_function">hashes</a>, and it’s often used to verify the integrity of files. For a given input, a <a href="https://realpython.com/python-hash-table/#understand-the-hash-function"><strong>hash function</strong></a> always returns the same value. Any minor changes in the input will result in a different hash value. Before you use the utility with concrete parameters, you may try to display the help:</p>

```python
! shasum --help
```

<p>Displaying the help of a command line program is a common feature exposed in the command-line interface.</p>

<p>To calculate the SHA-1 hash value of the content of a file, you proceed as follows:</p>

```python
! shasum main.c
```

<p>The result shows the SHA-1 hash value as the first field and the name of the file as the second field. The command can take more than one file as arguments:</p>

```python
! shasum main.c main.py
```

<p>Thanks to the wildcards expansion feature of the Unix terminal, it’s also possible to provide Python command-line arguments with wildcard characters. One such a character is the asterisk or star (<code>*</code>):</p>

```python
! shasum main.*
```

<p>The shell converts <code>main.*</code> to <code>main.c</code> and <code>main.py</code>, which are the two files matching the pattern <code>main.*</code> in the current directory, and passes them to <code>sha1sum</code>. The program calculates the <strong>SHA1 hash</strong> of each of the files in the argument list. You’ll see that, on Windows, the behavior is different. Windows has no wildcard expansion, so the program may have to accommodate for that. Your implementation may need to expand wildcards internally.</p>                                                                                                                                                                                                             
<p>Without any argument, <code>sha1sum</code> reads from the standard input. You can feed data to the program by typing characters on the keyboard. The input may incorporate any characters, including the carriage return <span class="keys"><kbd class="key-enter">Enter</kbd></span>. To terminate the input, you must signal the <a href="https://en.wikipedia.org/wiki/End-of-file">end of file</a> with <span class="keys"><kbd class="key-enter">Enter</kbd></span>, followed by the sequence <span class="keys"><kbd class="key-control">Ctrl</kbd><span>+</span><kbd class="key-d">D</kbd></span>:</p>

```python
! shasum
```

<p>You first enter the name of the program, <code>sha1sum</code>, followed by <span class="keys"><kbd class="key-enter">Enter</kbd></span>, and then <code>Real</code> and <code>Python</code>, each also followed by <span class="keys"><kbd class="key-enter">Enter</kbd></span>. To close the input stream, you type <span class="keys"><kbd class="key-control">Ctrl</kbd><span>+</span><kbd class="key-d">D</kbd></span>. The result is the value of the SHA1 hash generated for the text <code>Real\nPython\n</code>. The name of the file is <code>-</code>. This is a convention to indicate the standard input. The hash value is the same when you execute the following commands:</p>

```python
! python -c "print('Real\nPython\n', end='')" | shasum
```

```python
! python -c "print('Real\nPython')" | shasum
```

```python
! printf "Real\nPython\n" | shasum
```

## <h3 id="displaying-arguments">Displaying Arguments<a class="headerlink" href="#displaying-arguments"></a></h3>

<p>The <code>sys</code> module exposes an array named <code>argv</code> that includes the following:</p>

<ol>
<li><strong><code>argv[0]</code></strong> contains the name of the current Python program.</li>
<li><strong><code>argv[1:]</code></strong>, the rest of the list, contains any and all Python command-line arguments passed to the program.</li>
</ol>

<p>The following example demonstrates the content of <code>sys.argv</code>:</p>

```python
# argv.py
import sys

print(f"Name of the script      : {sys.argv[0]=}")
print(f"Arguments of the script : {sys.argv[1:]=}")
```

<p>Here’s how this code works:</p>

<ul>
<li><strong>Line 2</strong> imports the internal Python module <a href="https://docs.python.org/library/sys.html"><code>sys</code></a>.</li>
<li><strong>Line 4</strong> extracts the name of the program by accessing the first element of the list <code>sys.argv</code>.</li>
<li><strong>Line 5</strong> displays the Python command-line arguments by fetching all the remaining elements of the list <code>sys.argv</code>.</li>
</ul>

<div class="alert alert-block alert-info">
<p><strong>Note</strong>: The <a href="https://realpython.com/python-f-strings/">f-string</a> syntax used in <code>argv.py</code> leverages the new debugging specifier in Python 3.8. To read more about this new f-string feature and others, check out <a href="https://realpython.com/python38-new-features/">Cool New Features in Python 3.8</a>.</p>

<p>If your Python version is less than 3.8, then simply remove the equals sign (<code>=</code>) in both f-strings to allow the program to execute successfully. The output will only display the value of the variables, not their names.</p>
</div>

<p>Execute the script <code>argv.py</code> above with a list of arbitrary arguments as follows:</p>

```python
! python argv.py un deux trois quatre
```

<p>The output confirms that the content of <code>sys.argv[0]</code> is the Python script <code>argv.py</code>, and that the remaining elements of the <code>sys.argv</code> list contains the arguments of the script, <code>['un', 'deux', 'trois', 'quatre']</code>.</p>

<p>To summarize, <code>sys.argv</code> contains all the <code>argv.py</code> Python command-line arguments. When the Python interpreter executes a Python program, it parses the command line and populates <code>sys.argv</code> with the arguments.</p>

## <h3 id="reversing-the-first-argument">Reversing the First Argument<a class="headerlink" href="#reversing-the-first-argument"></a></h3>

<p>Now that you have enough background on <code>sys.argv</code>, you’re going to operate on arguments passed at the command line. The example <code>reverse.py</code> reverses the first argument passed at the command line:</p>

```python
# reverse.py

import sys

arg = sys.argv[1]
print(arg[::-1])
```

<p>In <code>reverse.py</code> the process to reverse the first argument is performed with the following steps:</p>

<ul>
<li><strong>Line 5</strong> fetches the first argument of the program stored at index <code>1</code> of <code>sys.argv</code>. Remember that the program name is stored at index <code>0</code> of <code>sys.argv</code>.</li>
<li><strong>Line 6</strong> prints the reversed string. <code>args[::-1]</code> is a Pythonic way to use a slice operation to <a href="https://stackoverflow.com/questions/3705670/best-way-to-create-a-reversed-list-in-python/3705676#3705676">reverse a list</a>.</li>
</ul>

<p>You execute the script as follows:</p>

```python
! python reverse.py "Real Python"
```

<p>As expected, <code>reverse.py</code> operates on <code>"Real Python"</code> and reverses the only argument to output <code>"nohtyP laeR"</code>. Note that surrounding the multi-word string <code>"Real Python"</code> with quotes ensures that the interpreter handles it as a unique argument, instead of two arguments. You’ll delve into <strong>argument separators</strong> in a later <a href="#escaping-whitespace-characters">section</a>.</p>

## <h3 id="mutating-sysargv">Mutating <code>sys.argv</code><a class="headerlink" href="#mutating-sysargv"></a></h3>

<p><code>sys.argv</code> is <strong>globally available</strong> to your running Python program. All modules imported during the execution of the process have direct access to <code>sys.argv</code>. This global access might be convenient, but <code>sys.argv</code> isn’t immutable. You may want to implement a more reliable mechanism to expose program arguments to different modules in your Python program, especially in a complex program with multiple files.</p>

<p>Observe what happens if you tamper with <code>sys.argv</code>:</p>

```python
# argv_pop.py

import sys

print(sys.argv)
sys.argv.pop()
print(sys.argv)
```

<p>You invoke <a href="https://docs.python.org/tutorial/datastructures.html#more-on-lists"><code>.pop()</code></a> to remove and return the last item in <code>sys.argv</code>.</p>

<p>Execute the script above:</p>

```python
! python argv_pop.py un deux trois quatre
```

<p>Notice that the fourth argument is no longer included in <code>sys.argv</code>.</p>

<p>In a short script, you can safely rely on the global access to <code>sys.argv</code>, but in a larger program, you may want to store arguments in a separate variable. The previous example could be modified as follows:</p>

```python
# argv_var_pop.py

import sys

print(sys.argv)
args = sys.argv[1:]
print(args)
sys.argv.pop()
print(sys.argv)
print(args)
```

<p>This time, although <code>sys.argv</code> lost its last element, <code>args</code> has been safely preserved. <code>args</code> isn’t global, and you can pass it around to parse the arguments per the logic of your program.
The Python package manager, <a href="https://realpython.com/courses/what-is-pip/"><code>pip</code></a>, uses this
<a href="https://github.com/pypa/pip/blob/ce46f8524e194ea81c47dbf8a547698f12e61329/src/pip/_internal/__init__.py#L53">approach</a>.
Here’s a short excerpt of the <code>pip</code> source code:</p>

```python
def main(args=None):
    if args is None:
        args = sys.argv[1:]
```

<p>In this snippet of code taken from the <a href="https://realpython.com/what-is-pip/"><code>pip</code></a> source code, <a href="https://realpython.com/python-main-function/"><code>main()</code></a> saves into <code>args</code> the slice of <code>sys.argv</code> that contains only the arguments and not the file name. <code>sys.argv</code> remains untouched, and <code>args</code> isn’t impacted by any inadvertent changes to <code>sys.argv</code>.</p>

## <h3 id="escaping-whitespace-characters">Escaping Whitespace Characters<a class="headerlink" href="#escaping-whitespace-characters"></a></h3>

<p>In the <code>reverse.py</code> example you saw <a href="#reversing-the-first-argument">earlier</a>, the first and only argument is <code>"Real Python"</code>, and the result is <code>"nohtyP laeR"</code>. The argument includes a whitespace separator between <code>"Real"</code> and <code>"Python"</code>, and it needs to be escaped.</p>

<p>On Linux, whitespaces can be escaped by doing one of the following:</p>

<ol>
<li><strong>Surrounding</strong> the arguments with single quotes (<code>'</code>)</li>
<li><strong>Surrounding</strong> the arguments with double quotes (<code>"</code>)</li>
<li><strong>Prefixing</strong> each space with a backslash (<code>\</code>)</li>
</ol>

<p>Without one of the escape solutions, <code>reverse.py</code> stores two arguments, <code>"Real"</code> in <code>sys.argv[1]</code> and <code>"Python"</code> in <code>sys.argv[2]</code>:</p>

```python
! python reverse.py Real Python
```

<p>The output above shows that the script only reverses <code>"Real"</code> and that <code>"Python"</code> is ignored. To ensure both arguments are stored, you’d need to surround the overall string with double quotes (<code>"</code>).</p>
<p>You can also use a backslash (<code>\</code>) to escape the whitespace:</p>

```python
! python reverse.py Real\ Python
```

<p>With the backslash (<code>\</code>), the command shell exposes a unique argument to Python, and then to <code>reverse.py</code>.</p>

<p>In Unix shells, the <a href="https://en.wikipedia.org/wiki/Internal_field_separator">internal field separator (IFS)</a> defines characters used as <strong>delimiters</strong>. The content of the shell variable, <code>IFS</code>, can be displayed by running the following command:</p>

```python
! printf "%q\n" "$IFS"
```

<p>From the result above, <code>' \t\n'</code>, you identify three delimiters:</p>

<ol>
<li><strong>Space</strong> (<code>' '</code>)</li>
<li><strong>Tab</strong> (<code>\t</code>)</li>
<li><strong>Newline</strong> (<code>\n</code>)</li>
</ol>

<p>Prefixing a space with a backslash (<code>\</code>) bypasses the default behavior of the space as a delimiter in the string <code>"Real Python"</code>. This results in one block of text as intended, instead of two.</p>

## <h3 id="handling-errors">Handling Errors<a class="headerlink" href="#handling-errors"></a></h3>

<p>Python command-line arguments are <strong>loose strings</strong>. Many things can go wrong, so it’s a good idea to provide the users of your program with some guidance in the event they pass incorrect arguments at the command line. For example, <code>reverse.py</code> expects one argument, and if you omit it, then you get an error:</p>

```python
! python reverse.py
```

<p>The Python <a href="https://realpython.com/python-exceptions/">exception</a> <code>IndexError</code> is raised, and the corresponding
<a href="https://realpython.com/python-traceback/">traceback</a> shows that the error is caused by the expression <code>arg = sys.argv[1]</code>.
The message of the exception is <code>list index out of range</code>. You didn’t pass an argument at the command line, so there’s nothing in the list <code>sys.argv</code> at index <code>1</code>.</p>

<p>This is a common pattern that can be addressed in a few different ways. For this initial example, you’ll keep it brief by including the expression <code>arg = sys.argv[1]</code> in a <code>try</code> block. Modify the code as follows:</p>

```python
# reverse_exc.py

import sys

try:
    arg = sys.argv[1]
except IndexError:
    raise SystemExit(f"Usage: {sys.argv[0]} <string_to_reverse>")
print(arg[::-1])
```

<p>The expression on line 4 is included in a <code>try</code> block. Line 8 raises the built-in exception
<a href="https://docs.python.org/library/exceptions.html#SystemExit"><code>SystemExit</code></a>.
If no argument is passed to <code>reverse_exc.py</code>, then the process exits with a status code of <code>1</code> after printing the usage.
Note the integration of <code>sys.argv[0]</code> in the error message. It exposes the name of the program in the usage message.
Now, when you execute the same program without any Python command-line arguments, you can see the following output:</p>

```python
! python reverse_exc.py
```

```python
! echo $?
```

<p><code>reverse.py</code> didn’t have an argument passed at the command line. As a result, the program raises <code>SystemExit</code> with an error message. This causes the program to exit with a status of <code>1</code>, which displays when you print the special variable
<a href="https://en.wikipedia.org/wiki/Bash_(Unix_shell)#Conditional_execution"><code>$?</code></a> with
<a href="https://en.wikipedia.org/wiki/Echo_%28command%29"><code>echo</code></a>.</p>

## <h3 id="calculating-the-sha1sum">Calculating the <code>sha1sum</code><a class="headerlink" href="#calculating-the-sha1sum"></a></h3>

<p>You’ll write another script to demonstrate that, on <a href="https://en.wikipedia.org/wiki/Unix-like">Unix-like</a> systems, Python command-line arguments are passed by bytes from the OS. This script takes a string as an argument and outputs the hexadecimal <a href="https://en.wikipedia.org/wiki/SHA-1">SHA-1</a> hash of the argument:</p>

```python
# sha1sum.py

import sys
import hashlib

data = sys.argv[1]
m = hashlib.sha1()
m.update(bytes(data, 'utf-8'))
print(m.hexdigest())
```

<p>This is loosely inspired by <code>sha1sum</code>, but it intentionally processes a string instead of the contents of a file.
In <code>sha1sum.py</code>, the steps to ingest the Python command-line arguments and to output the result are the following:</p>

<ul>
<li><strong>Line 6</strong> stores the content of the first argument in <code>data</code>.</li>
<li><strong>Line 7</strong> instantiates a SHA1 algorithm.</li>
<li><strong>Line 8</strong> updates the SHA1 hash object with the content of the first program argument. Note that <code>hash.update</code> takes a byte array as an argument, so it’s necessary to convert <code>data</code> from a string to a bytes array.</li>
<li><strong>Line 9</strong> prints a <a href="https://docs.python.org/library/hashlib.html#hashlib.hash.hexdigest">hexadecimal representation</a> of the SHA1 hash computed on line 8.</li>
</ul>

<p>When you run the script with an argument, you get this:</p>

```python
! python sha1sum.py "Real Python"
```

<p>For the sake of keeping the example short, the script <code>sha1sum.py</code> doesn’t handle missing Python command-line arguments.
Error handling could be addressed in this script the same way you did it in <code>reverse_exc.py</code>.</p>

<div class="alert alert-block alert-info">
<p><strong>Note:</strong> Checkout <a href="https://docs.python.org/library/hashlib.html"><code>hashlib</code></a> for more details about the hash functions available in the Python standard library.</p>
</div>

<p>From the <code>sys.argv</code> <a href="https://docs.python.org/library/sys.html#sys.argv">documentation</a>, you learn that in order to get the original bytes of the Python command-line arguments, you can use <a href="https://docs.python.org/library/os.html#os.fsencode"><code>os.fsencode()</code></a>. By directly obtaining the bytes from <code>sys.argv[1]</code>, you don’t need to perform the string-to-bytes conversion of <code>data</code>:</p>

```python
# sha1sum_bytes.py

import os
import sys
import hashlib

data = os.fsencode(sys.argv[1])
m = hashlib.sha1()
m.update(data)
print(m.hexdigest())
```

<p>The main difference between <code>sha1sum.py</code> and <code>sha1sum_bytes.py</code> are highlighted in the following lines:</p>

<ul>
<li><strong>Line 7</strong> populates <code>data</code> with the original bytes passed to the Python command-line arguments.</li>
<li><strong>Line 9</strong> passes <code>data</code> as an argument to <a href="https://docs.python.org/library/hashlib.html#hashlib.hash.update"><code>m.update()</code></a>, which receives a <a href="https://docs.python.org/glossary.html#term-bytes-like-object">bytes-like object</a>.</li>
</ul>

<p>Execute <code>sha1sum_bytes.py</code> to compare the output:</p>

```python
! python sha1sum_bytes.py "Real Python"
```

<p>The hexadecimal value of the SHA1 hash is the same as in the previous <code>sha1sum.py</code> example.</p>

# <h2 id="the-anatomy-of-python-command-line-arguments">The Anatomy of Python Command-Line Arguments<a class="headerlink" href="#the-anatomy-of-python-command-line-arguments"></a></h2>

<p>Now that you’ve explored a few aspects of Python command-line arguments, most notably <code>sys.argv</code>, you’re going to apply some of the standards that are regularly used by developers while implementing a command-line interface.</p>

<p>Python command-line arguments are a subset of the command-line interface. They can be composed of different types of arguments:</p>

<ol>
<li><strong>Options</strong> modify the behavior of a particular command or program.</li>
<li><strong>Arguments</strong> represent the source or destination to be processed.</li>
<li><strong>Subcommands</strong> allow a program to define more than one command with the respective set of options and arguments.</li>
</ol>

<p>Before you go deeper into the different types of arguments, you’ll get an overview of the accepted standards that have been guiding the design of the command-line interface and arguments. These have been refined since the advent of the
<a href="https://en.wikipedia.org/wiki/Computer_terminal">computer terminal</a> in the mid-1960s.</p>

## <h3 id="standards">Standards<a class="headerlink" href="#standards"></a></h3>

<p>A few available <strong>standards</strong> provide some definitions and guidelines to promote consistency for implementing commands and their arguments. These are the main UNIX standards and references:</p>

<ul>
<li><a href="http://pubs.opengroup.org/onlinepubs/9699919799/basedefs/V1_chap12.html">POSIX Utility Conventions</a></li>
<li><a href="https://www.gnu.org/prep/standards/standards.html#Command_002dLine-Interfaces">GNU Standards for Command Line Interfaces</a></li>
<li><a href="http://docopt.org/">docopt</a></li>
</ul>

<p>The standards above define guidelines and nomenclatures for anything related to programs and Python command-line arguments.
The following points are examples taken from those references:</p>

<ul>
<li><strong>POSIX</strong>:<ul>
<li>A program or utility is followed by options, option-arguments, and operands.</li>
<li>All options should be preceded with a hyphen or minus (<code>-</code>) delimiter character.</li>
<li>Option-arguments should not be optional.</li>
</ul>
</li>
<li><strong>GNU</strong>:<ul>
<li>All programs should support two standard options, which are <code>--version</code> and <code>--help</code>.</li>
<li>Long-named options are equivalent to the single-letter Unix-style options. An example is <code>--debug</code> and <code>-d</code>.</li>
</ul>
</li>
<li><strong>docopt</strong>:<ul>
<li>Short options can be stacked, meaning that <code>-abc</code> is equivalent to <code>-a -b -c</code>.</li>
<li>Long options can have arguments specified after a space or the equals sign (<code>=</code>). The long option <code>--input=ARG</code> is equivalent to <code>--input ARG</code>.</li>
</ul>
</li>
</ul>

<p>These standards define notations that are helpful when you describe a command. A similar notation can be used to display the usage of a particular command when you invoke it with the option <code>-h</code> or <code>--help</code>.</p>

<p>The GNU standards are very similar to the POSIX standards but provide some modifications and extensions. Notably, they add the <strong>long option</strong> that’s a fully named option prefixed with two hyphens (<code>--</code>). For example, to display the help, the regular option is <code>-h</code> and the long option is <code>--help</code>.</p>

<div class="alert alert-block alert-info">
<p><strong>Note:</strong> You don’t need to follow those standards rigorously. Instead, follow the conventions that have been used successfully for years since the advent of UNIX. If you write a set of utilities for you or your team, then ensure that you <strong>stay consistent</strong> across the different utilities.</p>
</div>

<p>In the following sections, you’ll learn more about each of the command line components, options, arguments, and sub-commands.</p>

# <h2 id="a-few-methods-for-parsing-python-command-line-arguments">A Few Methods for Parsing Python Command-Line Arguments<a class="headerlink" href="#a-few-methods-for-parsing-python-command-line-arguments"></a></h2>

<p>Now you’re going to explore a few approaches to apprehend options, option-arguments, and operands. This is done by <strong>parsing</strong> Python command-line arguments. In this section, you’ll see some concrete aspects of Python command-line arguments and techniques to handle them. <p>This will serve as a preparation for options involving modules in the standard libraries or from external libraries that you’ll learn about later in this tutorial.</p>

<p>For something uncomplicated, the following pattern, which doesn’t enforce ordering and doesn’t handle option-arguments, may be enough:</p>

```python
# cul.py

import sys

opts = [opt for opt in sys.argv[1:] if opt.startswith("-")]
args = [arg for arg in sys.argv[1:] if not arg.startswith("-")]

if "-c" in opts:
    print(" ".join(arg.capitalize() for arg in args))
elif "-u" in opts:
    print(" ".join(arg.upper() for arg in args))
elif "-l" in opts:
    print(" ".join(arg.lower() for arg in args))
else:
    raise SystemExit(f"Usage: {sys.argv[0]} (-c | -u | -l) <arguments>...")
```

<li><strong><code>-u</code></strong> to convert the arguments to uppercase</li>
<li><strong><code>-l</code></strong> to convert the argument to lowercase</li>
</ul>

<p>The code collects and separates the different argument types using <a href="https://realpython.com/list-comprehension-python/">list comprehensions</a>:</p>

<ul>
<li><strong>Line 5</strong> collects all the <strong>options</strong> by filtering on any Python command-line arguments starting with a hyphen (<code>-</code>).</li>
<li><strong>Line 6</strong> assembles the program <strong>arguments</strong> by filtering out the options.</li>
</ul>

## <h3 id="regular-expressions">Regular Expressions<a class="headerlink" href="#regular-expressions"></a></h3>

<p>You can use a <a href="https://realpython.com/regex-python/">regular expression</a> to enforce a certain order, specific options and option-arguments, or even the type of arguments. To illustrate the usage of a regular expression to parse Python command-line arguments, you’ll implement a Python version of <a href="https://en.wikipedia.org/wiki/Seq_%28Unix%29"><code>seq</code></a>, which is a program that prints a sequence of numbers. Following the docopt conventions, a specification for <code>seq.py</code> could be this:</p>

<pre><code>
Print integers from &lt;first&gt; to &lt;last&gt;, in steps of &lt;increment&gt;.

Usage:
  python seq.py --help
  python seq.py [-s SEPARATOR] <last>
  python seq.py [-s SEPARATOR] <first> <last>
  python seq.py [-s SEPARATOR] <first> <increment> <last>

Mandatory arguments to long options are mandatory for short options too.
  -s, --separator=STRING use STRING to separate numbers (default: \n)
      --help             display this help and exit

If &lt;first&gt; or &lt;increment&gt; are omitted, they default to 1. When &lt;first&gt; is larger than &lt;last&gt;, &lt;increment&gt;, if not set, defaults to -1. The sequence of numbers ends when the sum of the current number and &lt;increment&gt; reaches the limit imposed by &lt;last&gt;.
</code></pre>

<p>First, look at a regular expression that’s intended to capture the requirements above:</p>

```python
args_pattern = re.compile(
    r"""
    ^
    (
        (--(?P<HELP>help).*)|
        ((?:-s|--separator)\s(?P<SEP>.*?)\s)?
        ((?P<OP1>-?\d+))(\s(?P<OP2>-?\d+))?(\s(?P<OP3>-?\d+))?
    )
    $
""",
    re.VERBOSE,
)
```

<p>To experiment with the regular expression above, you may use the snippet recorded on <a href="https://regex101.com/r/Wh2Rce/3">Regular Expression 101</a>. The regular expression captures and enforces a few aspects of the requirements given for <code>seq</code>. In particular, the command may take:</p>

<ol>
<li><strong>A help option</strong>, in short (<code>-h</code>) or long format (<code>--help</code>), captured as a <a href="https://www.regular-expressions.info/named.html">named group</a> called <strong><code>HELP</code></strong></li>
<li><strong>A separator option</strong>, <code>-s</code> or <code>--separator</code>, taking an optional argument, and captured as named group called <strong><code>SEP</code></strong></li>
<li><strong>Up to three integer operands</strong>, respectively captured as <strong><code>OP1</code></strong>, <strong><code>OP2</code></strong>, and <strong><code>OP3</code></strong></li>
</ol>

<p>For clarity, the pattern <code>args_pattern</code> above uses the flag <a href="https://docs.python.org/library/re.html#re.VERBOSE"><code>re.VERBOSE</code></a> on line 11. This allows you to spread the regular expression over a few lines to enhance readability. The pattern validates the following:</p>

<ul>
<li><strong>Argument order</strong>: Options and arguments are expected to be laid out in a given order. For example, options are expected before the arguments.</li>
<li><strong>Option values</strong>**: Only <code>--help</code>, <code>-s</code>, or <code>--separator</code> are expected as options.</li>
<li><strong>Argument mutual exclusivity</strong>: The option <code>--help</code> isn’t compatible with other options or arguments.</li>
<li><strong>Argument type</strong>: Operands are expected to be positive or negative integers.</li>
</ul>

<p>For the regular expression to be able to handle these things, it needs to see all Python command-line arguments in one string. You can collect them using <a href="https://docs.python.org/3/library/stdtypes.html#str.join">str.join()</a>:</p>

```python
arg_line = " ".join(sys.argv[1:])
```

<p>This makes <code>arg_line</code> a string that includes all arguments, except the program name, separated by a space.</p>

<p>Given the pattern <code>args_pattern</code> above, you can extract the Python command-line arguments with the following function:</p>

```python
from typing import List, Dict

def parse(arg_line: str) -> Dict[str, str]:
    args: Dict[str, str] = {}
    if match_object := args_pattern.match(arg_line):
        args = {k: v for k, v in match_object.groupdict().items()
                if v is not None}
    return args
```

<p>The pattern is already handling the order of the arguments, mutual exclusivity between options and arguments, and the type of the arguments. <code>parse()</code> is applying <a href="https://docs.python.org/library/re.html#re.match"><code>re.match()</code></a> to the argument line to extract the proper values and store the data in a dictionary.</p>

<p>The <a href="https://realpython.com/courses/dictionaries-python/">dictionary</a> includes the names of each group as keys and their respective values. For example, if the <code>arg_line</code> value is <code>--help</code>, then the dictionary is <code>{'HELP': 'help'}</code>.
If <code>arg_line</code> is <code>-s T 10</code>, then the dictionary becomes <code>{'SEP': 'T', 'OP1': '10'}</code>. You can expand the code block below to see an implementation of <code>seq</code> with regular expressions.</p>

```python
# seq_regex.py

from typing import List, Dict
import re
import sys

USAGE = (
    f"Usage: {sys.argv[0]} [-s <separator>] [first [increment]] last"
)

args_pattern = re.compile(
    r"""
    ^
    (
        (--(?P<HELP>help).*)|
        ((?:-s|--separator)\s(?P<SEP>.*?)\s)?
        ((?P<OP1>-?\d+))(\s(?P<OP2>-?\d+))?(\s(?P<OP3>-?\d+))?
    )
    $
""",
    re.VERBOSE,
)

def parse(arg_line: str) -> Dict[str, str]:
    args: Dict[str, str] = {}
    if match_object := args_pattern.match(arg_line):
        args = {k: v for k, v in match_object.groupdict().items()
                if v is not None}
    return args

def seq(operands: List[int], sep: str = "\n") -> str:
    first, increment, last = 1, 1, 1
    if len(operands) == 1:
        last = operands[0]
    if len(operands) == 2:
        first, last = operands
        if first > last:
            increment = -1
    if len(operands) == 3:
        first, increment, last = operands
    last = last + 1 if increment > 0 else last - 1
    return sep.join(str(i) for i in range(first, last, increment))

def main() -> None:
    args = parse(" ".join(sys.argv[1:]))
    if not args:
        raise SystemExit(USAGE)
    if args.get("HELP"):
        print(USAGE)
        return
    operands = [int(v) for k, v in args.items() if k.startswith("OP")]
    sep = args.get("SEP", "\n")
    print(seq(operands, sep))

if __name__ == "__main__":
    main()
```
