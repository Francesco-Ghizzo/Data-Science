# <h2 id="regexes-in-python-and-their-uses">Regexes in Python and Their Uses<a class="headerlink" href="#regexes-in-python-and-their-uses" title="Permanent link"></a></h2>

<p>Imagine you have a string object <code>s</code>. Now suppose you need to write Python code to find out whether <code>s</code> contains the substring <code>'123'</code>. There are at least a couple ways to do this. You could use the <code>in</code> operator:</p>

```python
s = 'foo123bar'
'123' in s
```

<p>If you want to know not only <em>whether</em> <code>'123'</code> exists in <code>s</code> but also <em>where</em> it exists, then you can use <code>.find()</code> or <code>.index()</code>. Each of these returns the character position within <code>s</code> where the substring resides:</p>

```python
s = 'foo123bar'
s.find('123')
```

<p>In these examples, the matching is done by a straightforward character-by-character comparison. That will get the job done in many cases. But sometimes, the problem is more complicated than that.</p>

<p>For example, rather than searching for a fixed substring like <code>'123'</code>, suppose you wanted to determine whether a string contains <em>any</em> three consecutive decimal digit characters, as in the strings <code>'foo123bar'</code>, <code>'foo456bar'</code>, <code>'234baz'</code>, and <code>'qux678'</code>.</p>

<p>Strict character comparisons won’t cut it here. This is where regexes in Python come to the rescue.</p>

## <h3 id="a-very-brief-history-of-regular-expressions">A (Very Brief) History of Regular Expressions<a class="headerlink" href="#a-very-brief-history-of-regular-expressions" title="Permanent link"></a></h3>

<p>In 1951, mathematician Stephen Cole Kleene described the concept of a <a href="https://en.wikipedia.org/wiki/Regular_language">regular language</a>, a language that is recognizable by a finite automaton and formally expressible using <a href="https://en.wikipedia.org/wiki/Regular_expression">regular expressions</a>. In the mid-1960s, computer science pioneer <a href="https://en.wikipedia.org/wiki/Ken_Thompson">Ken Thompson</a>, one of the original designers of Unix, implemented pattern matching in the <a href="https://en.wikipedia.org/wiki/QED_(text_editor)">QED text editor</a> using Kleene’s notation.</p>

<p>Since then, regexes have appeared in many programming languages, editors, and other tools as a means of determining whether a string matches a specified pattern. Python, <a href="https://realpython.com/oop-in-python-vs-java/">Java</a>, and Perl all support regex functionality, as do most Unix tools and many text editors.</p>

## <h3 id="the-re-module">The <code>re</code> Module<a class="headerlink" href="#the-re-module" title="Permanent link"></a></h3>

<p>Regex functionality in Python resides in a module named <code>re</code>. The <code>re</code> module contains many useful functions and methods, most of which you’ll learn about in the next tutorial in this series.</p>

<p>For now, you’ll focus predominantly on one function, <code>re.search()</code>.</p>

<p class="h4 mt-4"><code>re.search(&lt;regex&gt;, &lt;string&gt;)</code></p>

<blockquote>
<p>Scans a string for a regex match.</p>
</blockquote>
                  
<p><code>re.search(&lt;regex&gt;, &lt;string&gt;)</code> scans <code>&lt;string&gt;</code> looking for the first location where the pattern <code>&lt;regex&gt;</code> matches. If a match is found, then <code>re.search()</code> returns a <strong>match object</strong>. Otherwise, it returns <a href="https://realpython.com/null-in-python/"><code>None</code></a>.</p>

<p><code>re.search()</code> takes an optional third <code>&lt;flags&gt;</code> argument that you’ll learn about at the end of this tutorial.</p>

One way to do this is to import the entire module and then use the module name as a prefix when calling the function:</p>

</code></pre>

For the sake of brevity, the <code>import re</code> statement will usually be omitted, but remember that it’s always necessary.</p>

## <h3 id="first-pattern-matching-example">First Pattern-Matching Example<a class="headerlink" href="#first-pattern-matching-example" title="Permanent link"></a></h3>

<p>Now that you know how to gain access to <code>re.search()</code>, you can give it a try:</p>

```python
s = 'foo123bar'

# One last reminder to import!
import re

re.search('123', s)
```

<p>Here, the search pattern <code>&lt;regex&gt;</code> is <code>123</code> and <code>&lt;string&gt;</code> is <code>s</code>. The returned match object appears on <strong>line 7</strong>. Match objects contain a wealth of useful information that you’ll explore soon.</p>

<p>For the moment, the important point is that <code>re.search()</code> did in fact return a match object rather than <code>None</code>. That tells you that it found a match. In other words, the specified <code>&lt;regex&gt;</code> pattern <code>123</code> is present in <code>s</code>.</p>

<p>A match object is <a href="https://realpython.com/python-data-types/#boolean-type-boolean-context-and-truthiness"><strong>truthy</strong></a>, so you can use it in a <a href="https://realpython.com/python-boolean/">Boolean context</a> like a conditional statement:</p>

```python
if re.search('123', s):
    print('Found a match.')
else:
    print('No match.')
```

<p>The interpreter displays the match object as <code>&lt;_sre.SRE_Match object; span=(3, 6), match='123'&gt;</code>. This contains some useful information.</p>

<p><code>span=(3, 6)</code> indicates the portion of <code>&lt;string&gt;</code> in which the match was found. This means the same thing as it would in <a href="https://realpython.com/python-strings/#string-slicing">slice notation</a>:</p>

```python
s[3:6]
```

<p>In this example, the match starts at character position <code>3</code> and extends up to but not including position <code>6</code>.</p>
<p><code>match='123'</code> indicates which characters from <code>&lt;string&gt;</code> matched.</p>

<p>This is a good start. But in this case, the <code>&lt;regex&gt;</code> pattern is just the plain string <code>'123'</code>. The pattern matching here is still just character-by-character comparison, pretty much the same as the <code>in</code> operator and <code>.find()</code> examples shown earlier. The match object helpfully tells you that the matching characters were <code>'123'</code>, but that’s not much of a revelation since those were exactly the characters you searched for.</p>

<p>You’re just getting warmed up.</p>

## <h3 id="python-regex-metacharacters">Python Regex Metacharacters<a class="headerlink" href="#python-regex-metacharacters" title="Permanent link"></a></h3>

<p>The real power of regex matching in Python emerges when <code>&lt;regex&gt;</code> contains special characters called <strong>metacharacters</strong>. These have a unique meaning to the regex matching engine and vastly enhance the capability of the search.</p>

<p>Consider again the problem of how to determine whether a string contains any three consecutive decimal digit characters.</p>

<p>In a regex, a set of characters specified in square brackets (<code>[]</code>) makes up a <strong>character class</strong>. This metacharacter sequence matches any single character that is in the class, as demonstrated in the following example:</p>

```python
s = 'foo123bar'
re.search('[0-9][0-9][0-9]', s)
```

<p><code>[0-9]</code> matches any single decimal digit character—any character between <code>'0'</code> and <code>'9'</code>, inclusive. The full expression <code>[0-9][0-9][0-9]</code> matches any sequence of three decimal digit characters. In this case, <code>s</code> matches because it contains three consecutive decimal digit characters, <code>'123'</code>.</p>

<p>These strings also match:</p>

```python
re.search('[0-9][0-9][0-9]', 'foo456bar')
re.search('[0-9][0-9][0-9]', '234baz')
re.search('[0-9][0-9][0-9]', 'qux678')
```

<p>On the other hand, a string that doesn’t contain three consecutive digits won’t match:</p>

```python
print(re.search('[0-9][0-9][0-9]', '12foo34'))
```

<p>With regexes in Python, you can identify patterns in a string that you wouldn’t be able to find with the <code>in</code> operator or with string methods.</p>

<p>Take a look at another regex metacharacter. The dot (<code>.</code>) metacharacter matches any character except a newline, so it functions like a wildcard:</p>

```python
s = 'foo123bar'
re.search('1.3', s)
```

<p>In the first example, the regex <code>1.3</code> matches <code>'123'</code> because the <code>'1'</code> and <code>'3'</code> match literally, and the <code>.</code> matches the <code>'2'</code>. Here, you’re essentially asking, “Does <code>s</code> contain a <code>'1'</code>, then any character (except a newline), then a <code>'3'</code>?” The answer is yes for <code>'foo123bar'</code> but no for <code>'foo13bar'</code>.</p>

<p>These examples provide a quick illustration of the power of regex metacharacters. Character class and dot are but two of the metacharacters supported by the <code>re</code> module. There are many more. Next, you’ll explore them fully.</p>

<p>The metacharacter sequences in this section try to match a single character from the search string. When the regex parser encounters one of these metacharacter sequences, a match happens if the character at the current parsing position fits the description that the sequence describes.</p>

<!-- Character Class -->
<p class="h4 mt-4"><code>[]</code></p>

<p>Characters contained in square brackets (<code>[]</code>) represent a <strong>character class</strong>—an enumerated set of characters to match from. A character class metacharacter sequence will match any single character contained in the class.</p>

```python
re.search('ba[artz]', 'foobazqux')
```

In the example, the regex <code>ba[artz]</code> matches both <code>'bar'</code> and <code>'baz'</code> (and would also match <code>'baa'</code> and <code>'bat'</code>).</p>

<p>A character class can also contain a range of characters separated by a hyphen (<code>-</code>), in which case it matches any single character within the range. For example, <code>[a-z]</code> matches any lowercase alphabetic character between <code>'a'</code> and <code>'z'</code>, inclusive:</p>

<p><code>[0-9]</code> matches any digit character:</p>

<p>In this case, <code>[0-9][0-9]</code> matches a sequence of two digits. The first portion of the string <code>'foo123bar'</code> that matches is <code>'12'</code>.</p>

<p><code>[0-9a-fA-F]</code> matches any <a href="https://en.wikipedia.org/wiki/Hexadecimal">hexadecimal</a> digit character:</p>

<p>Here, <code>[0-9a-fA-F]</code> matches the first hexadecimal digit character in the search string, <code>'a'</code>. </p>

<div class="alert alert-primary" role="alert">

</div>

<p>As you’ve seen, you can specify a range of characters in a character class by separating characters with a hyphen. What if you want the character class to include a literal hyphen character? You can place it as the first or last character or escape it with a backslash (<code>\</code>):</p>

<p>If you want to include a literal <code>']'</code> in a character class, then you can place it as the first character or escape it with backslash:</p>

<p>As you saw in the table above, <code>*</code> and <code>+</code> have special meanings in a regex in Python. They designate repetition, which you’ll learn more about shortly. But in this example, they’re inside a character class, so they match themselves literally.</p>

<!-- Wildcard -->
<p>dot (<code>.</code>)</p>

<blockquote>
<p>Specifies a wildcard.</p>
</blockquote>

<p>The <code>.</code> metacharacter matches any single character except a newline:</p>

<p>As a regex, <code>foo.bar</code> essentially means the characters <code>'foo'</code>, then any character except newline, then the characters <code>'bar'</code>. The first string shown above, <code>'fooxbar'</code>, fits the bill because the <code>.</code> metacharacter matches the <code>'x'</code>.</p>

<blockquote>

</blockquote>

<p><code>\w</code> matches any alphanumeric word character. Word characters are uppercase and lowercase letters, digits, and the underscore (<code>_</code>) character, so <code>\w</code> is essentially shorthand for <code>[a-zA-Z0-9_]</code>:</p>

<p>In this case, the first word character in the string <code>'#(.a$@&amp;'</code> is <code>'a'</code>.</p>

<p><code>\W</code> is the opposite. It matches any non-word character and is equivalent to <code>[^a-zA-Z0-9_]</code>:</p>

<p>Here, the first non-word character in <code>'a_1*3!b'</code> is <code>'*'</code>.</p>

<!-- Digit Character -->
<p class="h4 mt-4"><code>\d</code><br><code>\D</code></p>

<blockquote>
<p>Match based on whether a character is a decimal digit.</p>
</blockquote>

<p><code>\d</code> matches any decimal digit character. <code>\D</code> is the opposite. It matches any character that <em>isn’t</em> a decimal digit:</p>

<p><code>\d</code> is essentially equivalent to <code>[0-9]</code>, and <code>\D</code> is equivalent to <code>[^0-9]</code>.</p>

<!-- Whitespace Character -->
<p class="h4 mt-4"><code>\s</code><br><code>\S</code></p>

<blockquote>
<p>Match based on whether a character represents whitespace.</p>
</blockquote>

<p><code>\s</code> matches any whitespace character:</p>

<p>Note that, unlike the dot wildcard metacharacter, <code>\s</code> does match a newline character.</p>

<p><code>\S</code> is the opposite of <code>\s</code>. It matches any character that <em>isn’t</em> whitespace:</p>

<p>Again, <code>\s</code> and <code>\S</code> consider a newline to be whitespace. In the example above, the first non-whitespace character is <code>'f'</code>.</p>

<p>The character class sequences <code>\w</code>, <code>\W</code>, <code>\d</code>, <code>\D</code>, <code>\s</code>, and <code>\S</code> can appear inside a square bracket character class as well:</p>

<p>In this case, <code>[\d\w\s]</code> matches any digit, word, or whitespace character. And since <code>\w</code> includes <code>\d</code>, the same character class could also be expressed slightly shorter as <code>[\w\s]</code>.</p>

## <h3 id="escaping-metacharacters">Escaping Metacharacters<a class="headerlink" href="#escaping-metacharacters" title="Permanent link"></a></h3>

<p>Occasionally, you’ll want to include a metacharacter in your regex, except you won’t want it to carry its special meaning. Instead, you’ll want it to represent itself as a literal character.</p>

<!-- Escape Character -->
<p class="h4 mt-4">backslash (<code>\</code>)</p>

<blockquote>
<p>Removes the special meaning of a metacharacter.</p>
</blockquote>

<p>As you’ve just seen, the backslash character can introduce special character classes like word, digit, and whitespace. There are also special metacharacter sequences called <strong>anchors</strong> that begin with a backslash, which you’ll learn about below.</p>

<p>When it’s not serving either of these purposes, the backslash <strong>escapes</strong> metacharacters. A metacharacter preceded by a backslash loses its special meaning and matches the literal character instead. Consider the following examples:</p>

<p>The <a href="https://en.wikipedia.org/wiki/Unicode_Consortium">Unicode Consortium</a> created Unicode to handle this problem. Unicode is a character-encoding standard designed to represent all the world’s writing systems. All strings in Python 3, including regexes, are Unicode by default.</p>
