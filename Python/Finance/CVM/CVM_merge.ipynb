# Imports

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_rows', 10)
```

# Constants

```python
CD_CONTA = 'Código da Conta'
DS_CONTA = 'Descrição da Conta'
```

# User defined functions

```python
def get_len_key(df):
    return [len(key) for key in df[CD_CONTA]]
```

```python
def get_min_len(len_key):
    return np.array(len_key).min()
```

```python
def get_ind_min_len(df, min_len):
    len_key = get_len_key(df)
    return np.where(len_key==min_len)
```

```python
def get_drop_mask(merge_df, len_key, min_len):
    can_drop = (merge_df[CD_CONTA] == merge_df.shift(-1)[CD_CONTA])
    return (len_key == min_len) & (can_drop == True)
```

# Read from file

```python
try:
    df = pd.read_csv('Financial_Statement.csv', sep=';')
except Exception as e:
    print(f"Error: {e}")
```

```python
df
```

# Wrangling

## Round 1

```python
round = 1
```

```python
df
```

### Merge

```python
len_key = get_len_key(df)
min_len = get_min_len(len_key)
ind_min_len = get_ind_min_len(df, min_len)  # ind_min_len = index of key 'Código da Conta' with minimum length
```

```python
df.iloc[ind_min_len]
```

```python
merge_keys = df[CD_CONTA].apply(lambda string: string[:min_len])
merge_df = pd.merge(merge_keys, df[[CD_CONTA, DS_CONTA]], how='left')
```

```python
merge_df
```

```python
df.insert(round, f'{DS_CONTA} {round}', merge_df[DS_CONTA])
```

```python
df
```

### Drop

```python
drop_mask = get_drop_mask(merge_df, len_key, min_len)
```

```python
df[drop_mask]
```

```python
df = df.drop(df[drop_mask].index)
df = df.reset_index(drop=True)
```
