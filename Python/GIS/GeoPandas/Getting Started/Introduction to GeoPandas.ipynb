<h1>Introduction to GeoPandas</h1>

<p>This quick tutorial introduces the key concepts and basic features of GeoPandas to help you get started with your projects.</p>

# <h2>Concepts</h2>

<p>GeoPandas, as the name suggests, extends the popular data science library <a href="https://pandas.pydata.org">pandas</a> by adding support for geospatial data. If you are not familiar with <code>pandas</code>, we recommend taking a quick look at its
<a href="https://pandas.pydata.org/docs/getting_started/index.html#getting-started">Getting started documentation</a> before proceeding.</p>

<p>The core data structure in GeoPandas is the <code>geopandas.GeoDataFrame</code>, a subclass of <code>pandas.DataFrame</code>, that can store geometry columns and perform spatial operations.
The <code>geopandas.GeoSeries</code>, a subclass of <code>pandas.Series</code>, handles the geometries.
Therefore, your <code>GeoDataFrame</code> is a combination of <code>pandas.Series</code>, with traditional data (numerical, boolean, text etc.), and <code>geopandas.GeoSeries</code>, with geometries (points, polygons etc.).
You can have as many columns with geometries as you wish; unlike in some typical desktop GIS software.</p>

<p><img src="https://geopandas.org/en/stable/_images/dataframe.svg"></p>

<p>Each <code>GeoSeries</code> can contain any geometry type (you can even mix them within a single array) and has a <code>GeoSeries.crs</code> attribute, which stores information about the projection (CRS stands for Coordinate Reference System). Therefore, each <code>GeoSeries</code> in a <code>GeoDataFrame</code> can be in a different projection, allowing you to have, for example, multiple versions (different projections) of the same geometry.</p>

<p>Only one <code>GeoSeries</code> in a <code>GeoDataFrame</code> is considered the <em>active geometry column</em>, which means that all geometric operations applied to a <code>GeoDataFrame</code> operate on this column. The active geometry column is accessed via the <code>GeoDataFrame.geometry</code> attribute.</p>

# <h2>Reading and writing files</h2>

## <h3>Reading files</h3>

<p>Assuming you have a file containing both data and geometry (e.g. GeoPackage, GeoJSON, Shapefile), you can read it using <code>geopandas.read_file()</code>, which automatically detects the filetype and creates a <code>GeoDataFrame</code>. This tutorial uses the <code>"nybb"</code> dataset, a map of New York boroughs, which is available through the <code>geodatasets</code> package. Therefore, we use <code>geodatasets.get_path()</code> to download the dataset and retrieve the path to the local copy.</p>

<div class="alert alert-block alert-warning">
<p>Warning</p>
<p>For operations that rely on distance or area, you always need to use a projected CRS (in meters, feet, kilometers etc.) not a geographic one (in degrees).
