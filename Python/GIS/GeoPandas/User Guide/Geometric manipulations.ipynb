<h1>Geometric manipulations</h1>

<p>GeoPandas makes available all the tools for geometric manipulations in the <a href="http://shapely.readthedocs.io/en/latest/manual.html">Shapely library</a>.</p>

<p>Note that documentation for all set-theoretic tools for creating new shapes using the relationship between two different spatial datasets – like creating intersections, or differences – can be found at <a href="https://geopandas.org/en/stable/docs/user_guide/set_operations.html">Set operations with overlay</a>.</p>

# <h2>Constructive methods</h2>

<dl>
<dt id="GeoSeries.buffer">
GeoSeries.buffer(<em>distance</em>, <em>resolution=16</em>)</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of geometries representing all points within a given <cite>distance</cite> of each geometric object.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.boundary">
GeoSeries.boundary</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of lower dimensional objects representing each geometry’s set-theoretic <cite>boundary</cite>.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.centroid">
GeoSeries.centroid</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of points for each geometric centroid.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.concave_hull">
GeoSeries.concave_hull</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of geometries representing the smallest concave <cite>Polygon</cite> containing all the points in each object unless the number of points in the object is less than three. For two points, the concave hull collapses to a <cite>LineString</cite>; for 1, a <cite>Point</cite>.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.convex_hull">
GeoSeries.convex_hull</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of geometries representing the smallest convex <cite>Polygon</cite> containing all the points in each object unless the number of points in the object is less than three. For two points, the convex hull collapses to a <cite>LineString</cite>; for 1, a <cite>Point</cite>.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.constrained_delaunay_triangles">
GeoSeries.constrained_delaunay_triangles()</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> with the constrained Delaunay triangulation of polygons. A constrained Delaunay triangulation requires the edges of the input polygon(s) to be in the set of resulting triangle edges. An unconstrained delaunay triangulation only triangulates based on the vertices, hence triangle edges could cross polygon boundaries.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.delaunay_triangles">
GeoSeries.delaunay_triangles(<em>tolerance</em>, <em>preserve_topology=True</em>)</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> consisting of polygons (default) or linestrings (<cite>only_edges=True</cite>) representing the computed Delaunay triangulation around the vertices of an input geometry.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.envelope">
GeoSeries.envelope</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of geometries representing the point or smallest rectangular polygon (with sides parallel to the coordinate axes) that contains each object.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.extract_unique_points">
GeoSeries.extract_unique_points()</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of geometries containing all distinct vertices of each input geometry as a multipoint.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.offset_curve">
GeoSeries.offset_curve(<em>distance</em>, <em>quad_segs=8</em>, <em>join_style='round'</em>, <em>mitre_limit=5.0</em>)</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> containing a <cite>Linestring</cite> or <cite>MultiLineString</cite> geometry at a distance from the object on its right or its left side.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.remove_repeated_points">
GeoSeries.remove_repeated_points()</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> containing a copy of the input geometry with repeated points removed.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.simplify">

<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> containing a simplified representation of
each object.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.segmentize">
GeoSeries.segmentize(<em>max_segment_length</em>)</dt>
<dd><p>Returns a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> with additional vertices added to line segments based on max_segment_length.</p>
</dd></dl>

<dl>
<dt id="GeoSeries.union_all">
GeoSeries.union_all()</dt>
<dd><p>Return a geometry containing the union of all geometries in the

</dd></dl>

# <h2>Examples of geometric manipulations</h2>

```python
import geopandas
from geopandas import GeoSeries
from shapely.geometry import Polygon
```

```python
p1 = Polygon([(0, 0), (1, 0), (1, 1)])
p2 = Polygon([(0, 0), (1, 0), (1, 1), (0, 1)])
p3 = Polygon([(2, 0), (3, 0), (3, 1), (2, 1)])
```

```python
g = GeoSeries([p1, p2, p3])
```

```python
g
```

<p>GeoPandas also implements alternate constructors that can read any data format recognized by
<a href="http://pyogrio.readthedocs.io/en/latest/">Pyogrio</a>. To read a zip file containing an ESRI shapefile with the
<a href="https://data.cityofnewyork.us/City-Government/Borough-Boundaries/tqmj-j8zm">borough boundaries of New York City</a> (provided by the <code>geodatasets</code> package):</p>

```python
import geodatasets

nybb_path = geodatasets.get_path('nybb')
boros = geopandas.read_file(nybb_path)
boros.set_index('BoroCode', inplace=True)
boros.sort_index(inplace=True)
```

```python
boros
```

```python
boros['geometry'].convex_hull
```

<p>To demonstrate a more complex operation, generate a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> containing 2000 random points:</p>

```python
import numpy as np
from shapely.geometry import Point
```

```python
xmin, xmax, ymin, ymax = 900000, 1080000, 120000, 280000
xc = (xmax - xmin) * np.random.random(2000) + xmin
yc = (ymax - ymin) * np.random.random(2000) + ymin
pts = GeoSeries([Point(x, y) for x, y in zip(xc, yc)])
```

<p>Now draw a circle with fixed radius around each point:</p>

```python
circles = pts.buffer(2000)
```

<p>You can collapse these circles into a single <a href="https://shapely.readthedocs.io/en/stable/manual.html#MultiPolygon" title="(in Shapely)"><code>MultiPolygon</code></a> geometry with</p>

```python
mp = circles.union_all()
```

<p>To extract the part of this geometry contained in each borough, you can just use:</p>

```python
holes = boros['geometry'].intersection(mp)
```

<img src="https://geopandas.org/en/stable/_images/holes.png">

<p>and to get the area outside of the holes:</p>

```python
boros_with_holes = boros['geometry'].difference(mp)
```

<img src="https://geopandas.org/en/stable/_images/boros_with_holes.png">

<p>Note that this can be simplified a bit, since <code>geometry</code> is available as an attribute on a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a>, and the
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.intersection.html" title="geopandas.GeoSeries.intersection"><code>intersection()</code></a> and
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.difference.html" title="geopandas.GeoSeries.difference"><code>difference()</code></a> methods are implemented with the “&amp;” and “-” operators, respectively.
For example, the latter could have been expressed simply as <code>boros.geometry - mp</code>.</p>

<p>It’s easy to do things like calculate the fractional area in each borough that are in the holes:</p>

```python
holes.area / boros.geometry.area
```
