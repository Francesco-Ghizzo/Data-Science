<h1>Indexing and selecting data</h1>

<p>GeoPandas inherits the standard <a href="http://pandas.pydata.org/pandas-docs/stable/indexing.html">pandas</a> methods for indexing/selecting data. This includes label based indexing with <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.loc.html" title="(in pandas v2.3.0)"><code>loc</code></a> and integer position based indexing with <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.iloc.html" title="(in pandas v2.3.0)"><code>iloc</code></a>, which apply to both
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> and
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> objects. For more information on indexing/selecting, see the <a href="http://pandas.pydata.org/pandas-docs/stable/indexing.html">pandas</a> documentation.</p>

<p>In addition to the standard <a href="http://pandas.pydata.org/pandas-docs/stable/indexing.html">pandas</a> methods, GeoPandas also provides
coordinate based indexing with the <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.cx.html" title="geopandas.GeoDataFrame.cx"><code>cx</code></a> indexer, which slices using a bounding box.
Geometries in the <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</span></code></a> or <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> that intersect the bounding box will be returned.</p>

<p>Using the <code>geoda.chile_labor</code> dataset, you can use this functionality to quickly select parts of Chile whose boundaries extend south of the -50 degrees latitude. You can first check the original GeoDataFrame.</p>

```python
import geopandas, geodatasets
```

```python
chile = geopandas.read_file(geodatasets.get_path('geoda.chile_labor'))
```

```python
chile.plot(figsize=(8, 8))
```

<p>And then select only the southern part of the country.</p>

```python
southern_chile = chile.cx[:, :-50]
```

```python
southern_chile.plot(figsize=(8, 8))
```
