<h1>Data structures</h1>

<p>GeoPandas implements two main data structures, a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> and a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a>. These are subclasses of <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="(in pandas v2.3.0)"><code>pandas.Series</code></a> and
<a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.html" title="(in pandas v2.3.0)"><code>pandas.DataFrame</code></a>, respectively.</p>

# <h2>GeoSeries</h2>

<p>A <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> is essentially a vector where each entry in the vector is a set of shapes corresponding to one observation.
An entry may consist of only one shape (like a single polygon) or multiple shapes that are meant to be thought of as one observation (like the many polygons that make up the State of Hawaii or a country like Indonesia).</p>

<p>GeoPandas has three basic classes of geometric objects (which are actually
<a href="https://shapely.readthedocs.io/en/stable/manual.html">Shapely</a> objects):</p>

<ul class="simple">
<li><p>Points / Multi-Points</p></li>
<li><p>Lines / Multi-Lines</p></li>
<li><p>Polygons / Multi-Polygons</p></li>
</ul>

<p>Note that all entries in a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> do not need to be of the same geometric type, although certain export operations will fail if this is not the case.</p>

## <h3>Overview of attributes and methods</h3>

<p>The <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> class implements nearly all of the attributes and methods of Shapely objects.
When applied to a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a>, they will apply elementwise to all geometries in the series.
Binary operations can be applied between two <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a>, in which case the operation is carried out elementwise.  The two series will be aligned
by matching indices.  Binary operations can also be applied to a single geometry, in which case the operation is carried out for each
element of the series with that geometry.  In either case, a
<a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="(in pandas v2.3.0)"><code>Series</code></a> or a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> will be returned, as appropriate.</p>

<p>A short summary of a few attributes and methods for GeoSeries is presented here, and a full list can be found in the
<a href="https://geopandas.org/en/stable/docs/reference/geoseries.html">GeoSeries API reference</a>.
There is also a family of methods for creating new shapes by expanding existing shapes or applying set-theoretic operations like “union” described
in <a href="geometric_manipulations.html">Geometric manipulations</a>.</p>

### <h4>Attributes</h4>

<ul class="simple">
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.area.html" title="geopandas.GeoSeries.area"><code>area</code></a>: shape area (units of projection – see <a href="https://geopandas.org/en/stable/docs/user_guide/projections.html">projections</a>)</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.bounds.html" title="geopandas.GeoSeries.bounds"><code>bounds</code></a>: tuple of max and min coordinates on each axis for each shape</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.total_bounds.html" title="geopandas.GeoSeries.total_bounds"><code>total_bounds</code></a>: tuple of max and min coordinates on each axis for entire GeoSeries</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.geom_type.html" title="geopandas.GeoSeries.geom_type"><code>geom_type</code></a>: type of geometry.</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.is_valid.html" title="geopandas.GeoSeries.is_valid"><code>is_valid</code></a>: tests if coordinates make a shape that is reasonable geometric shape according to the
<a href="http://www.opengeospatial.org/standards/sfa">Simple Feature Access</a> standard.</p></li>
</ul>

### <h4>Basic methods</h4>

<ul class="simple">
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.distance.html" title="geopandas.GeoSeries.distance"><code>distance()</code></a>: returns <a href="https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.html" title="(in pandas v2.3.0)"><code>Series</code></a> with minimum distance from each entry to <code>other</code></p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/user_guide/geometric_manipulations.html#GeoSeries.centroid" title="GeoSeries.centroid"><code>centroid</code></a>: returns <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of centroids</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.representative_point.html" title="geopandas.GeoSeries.representative_point"><code>representative_point()</code></a>:  returns
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> of points that are guaranteed to be within each geometry. It does <strong>NOT</strong> return centroids.</p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.to_crs.html" title="geopandas.GeoSeries.to_crs"><code>to_crs()</code></a>: change coordinate reference system.
See <a href="https://geopandas.org/en/stable/docs/user_guide/projections.html">projections</a></p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.plot.html" title="geopandas.GeoSeries.plot"><code>plot()</code></a>: plot <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a>. See <a href="https://geopandas.org/en/stable/docs/user_guide/mapping.html">mapping</a>.</p></li>
</ul>

# <h2>GeoDataFrame</h2>

<p>A <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> is a tabular data structure that contains a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a>.</p>

<p>The most important property of a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> is that it always has one <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> column that holds a special status - the “active geometry column”. When a spatial method is applied to a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> (or a spatial attribute like <code>area</code> is called), these operations will always act on the active geometry column.</p>

<p>The active geometry column – no matter the name of the corresponding
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a> –
can be accessed through the <code>geometry</code> attribute (<code>gdf.geometry</code>), and the name of the <code>geometry</code> column can be found by typing <code>gdf.geometry.name</code> or <code>gdf.active_geometry_name</code>.</p>

<p>A <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> may also contain other columns with geometrical (shapely) objects, but only one column can be the active geometry at a time. To change which column is the active geometry column, use the
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.set_geometry.html" title="geopandas.GeoDataFrame.set_geometry"><code>GeoDataFrame.set_geometry()</code></a> method.</p>

<p><strong>Note 2:</strong> Somewhat confusingly, by default when you use the
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.read_file.html" title="geopandas.read_file"><code>read_file()</code></a> command, the column containing spatial objects from the file is named “geometry” by default, and will be set as the active geometry column.
However, despite using the same term for the name of the column and the name of the special attribute that keeps track of the active column, they are distinct. You can easily shift the active geometry column to a different
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</span></code></a> with the
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.set_geometry.html" title="geopandas.GeoDataFrame.set_geometry"><code>set_geometry()</code></a> command. Further, <code>gdf.geometry</code> will always return the active geometry column, <em>not</em> the column named <code>geometry</code>. If you wish to call a column named “geometry”, and a different column is the active geometry column, use <code>gdf['geometry']</code>, not <code>gdf.geometry</code>.</p>

## <h3>Attributes and methods</h3>

<p>Any of the attributes calls or methods described for a <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</span></code></a> will work on a
<a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoDataFrame.html" title="geopandas.GeoDataFrame"><code>GeoDataFrame</code></a> – they are just applied to the active geometry column <a href="https://geopandas.org/en/stable/docs/reference/api/geopandas.GeoSeries.html" title="geopandas.GeoSeries"><code>GeoSeries</code></a>.</p>

<p>However, <a href="" title="geopandas.GeoDataFrame"><code>GeoDataFrames</code></a> also have a number few extra methods for:</p>

<ul class="simple">
<li><p><a href="https://geopandas.org/en/stable/docs/user_guide/io.html">Reading and writing files</a></p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/user_guide/mergingdata.html#mergingdata-spatial-joins">Spatial joins</a></p></li>
<li><p><a href="https://geopandas.org/en/stable/docs/user_guide/aggregation_with_dissolve.html">Spatial aggregations</a></p></li>

</ul>
