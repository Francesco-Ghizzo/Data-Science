<h1>Keep or drop columns using their names and types</h1>

<small>Source: <a href="https://github.com/tidyverse/dplyr/blob/pkgdown-v1.1.4/R/select.R"><code>R/select.R</code></a></small>
<div><code>select.Rd</code></div>

<p>Select (and optionally rename) variables in a data frame, using a concise mini-language that makes it easy to refer to variables based on their name (e.g. <code>a:f</code> selects all columns from <code>a</code> on the left to <code>f</code> on the right) or type (e.g. <code>where(is.numeric)</code> selects all numeric columns).</p>

## <h3 id="overview-of-selection-features">Overview of selection features</h3>

<p>Tidyverse selections implement a dialect of R where operators make
it easy to select variables:</p>

<ul>
<li><p><code>:</code> for selecting a range of consecutive variables.</p></li>
<li><p><code>!</code> for taking the complement of a set of variables.</p></li>
<li><p><code>&amp;</code> and <code>|</code> for selecting the intersection or the union of two sets of variables.</p></li>
<li><p><code><a href="https://rdrr.io/r/base/c.html">c()</a></code> for combining selections.</p></li>
</ul>

<p>In addition, you can use <strong>selection helpers</strong>. Some helpers select specific columns:</p>

<ul>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/everything.html">everything()</a></code>: Matches all variables.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/everything.html">last_col()</a></code>: Select last variable, possibly with an offset.</p></li>
<li><p><code><a href="group_cols.html">group_cols()</a></code>: Select all grouping columns.</p></li>
</ul>

<p>Other helpers select variables by matching patterns in their names:</p>

<ul>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">starts_with()</a></code>: Starts with a prefix.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">ends_with()</a></code>: Ends with a suffix.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">contains()</a></code>: Contains a literal string.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">matches()</a></code>: Matches a regular expression.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">num_range()</a></code>: Matches a numerical range like x01, x02, x03.</p></li>
</ul><p>Or from variables stored in a character vector:</p><ul><li><p><code><a href="https://tidyselect.r-lib.org/reference/all_of.html">all_of()</a></code>: Matches variable names in a character vector. All names must be present, otherwise an out-of-bounds error is thrown.</p></li>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/all_of.html">any_of()</a></code>: Same as <code><a href="https://tidyselect.r-lib.org/reference/all_of.html">all_of()</a></code>, except that no error is thrown for names that don't exist.</p></li>
</ul>

<p>Or using a predicate function:</p>

<ul>
<li><p><code><a href="https://tidyselect.r-lib.org/reference/where.html">where()</a></code>: Applies a function to all variables and selects those for which the function returns <code>TRUE</code>.</p></li>
</ul>

# <h2 id="ref-usage">Usage</h2>

<pre class="sourceCode r"><code>select(.data, ...)</code></pre>

# <h2 id="arguments">Arguments</h2>

<dl>

<dt id="arg--data">.data</dt>
<dd><p>A data frame, data frame extension (e.g. a tibble), or a lazy data frame (e.g. from dbplyr or dtplyr). See <em>Methods</em>, below, for more details.</p></dd>

<dt id="arg--">...</dt>
<dd><p>&lt;<code><a href="dplyr_tidy_select.html">tidy-select</a></code>&gt; One or more unquoted expressions separated by commas. Variable names can be used as if they were positions in the data frame, so expressions like <code>x:y</code> can be used to select a range of variables.</p></dd>

</dl>

# <h2 id="value">Value</h2>

<p>An object of the same type as <code>.data</code>. The output has the following properties:</p>

<ul>
<li><p>Rows are not affected.</p></li>
<li><p>Output columns are a subset of input columns, potentially with a different order. Columns will be renamed if <code>new_name = old_name</code> form is used.</p></li>
<li><p>Data frame attributes are preserved.</p></li>
<li><p>Groups are maintained; you can't select off grouping variables.</p></li>
</ul>

# <h2 id="examples">Examples</h2>

<p>Here we show the usage for the basic selection operators. See the
specific help pages to learn about helpers like <code><a href="https://tidyselect.r-lib.org/reference/starts_with.html">starts_with()</a></code>.</p>

<p>The selection language can be used in functions like
<code>dplyr::select()</code> or <code><a href="https://tidyr.tidyverse.org/reference/pivot_longer.html">tidyr::pivot_longer()</a></code>. Let's first attach the tidyverse:</p>

```r
library(tidyverse)

# For better printing
iris <- as_tibble(iris)
```

<p>Select variables by name:</p>

```r
starwars |> select(height)
```

```r
iris |> pivot_longer(Sepal.Length)
```

<p>Select multiple variables by separating them with commas. Note how
the order of columns is determined by the order of inputs:</p>

```r
starwars |> select(homeworld, height, mass)
```

<p>Functions like <code><a href="https://tidyr.tidyverse.org/reference/pivot_longer.html">tidyr::pivot_longer()</a></code> don't take variables with dots. In this case use <code><a href="https://rdrr.io/r/base/c.html">c()</a></code> to select multiple variables:</p>

```r
iris |> pivot_longer(c(Sepal.Length, Petal.Length))
```

## <h3 id="operators-">Operators:</h3>

<p>The <code>:</code> operator selects a range of consecutive variables:</p>

```r
starwars |> select(name:mass)
```

<p>The <code>!</code> operator negates a selection:</p>

```r
starwars |> select(!(name:mass))
```

```r
iris |> select(!c(Sepal.Length, Petal.Length))
```

```r
iris |> select(!ends_with("Width"))
```

<p><code>&amp;</code> and <code>|</code> take the intersection or the union of two selections:</p>

```r
iris |> select(starts_with("Petal") & ends_with("Width"))
```

```r
iris |> select(starts_with("Petal") | ends_with("Width"))
```

<p>To take the difference between two selections, combine the <code>&amp;</code> and <code>!</code> operators:</p>

```r
iris |> select(starts_with("Petal") & !ends_with("Width"))
```
